```python

import os
import tarfile
import subprocess
from pathlib import Path

import torch
import torchvision
from sklearn.model_selection import train_test_split

print("GPU:", torch.cuda.get_device_name(0))

# Download official Imagenette 320 archive
URL = "https://s3.amazonaws.com/fast-ai-imageclas/imagenette2-320.tgz"
ARCHIVE = "/content/imagenette2-320.tgz"
EXTRACT = Path("/content/imagenette2-320")

if not EXTRACT.exists():
    print("Downloading Imagenette 320...")
    subprocess.run(
        ["wget", "-q", "--show-progress", URL, "-O", ARCHIVE],
        check=True
    )
    print("Extracting dataset...")
    with tarfile.open(ARCHIVE, "r:gz") as tar:
        tar.extractall("/content")
else:
    print("Imagenette archive already extracted.")

# Inspect official dataset folders
ROOT = EXTRACT

train_root = ROOT / "train"
val_root = ROOT / "val"

assert train_root.exists(), "Imagenette train directory missing"
assert val_root.exists(), "Imagenette validation directory missing"

classes = sorted([p.name for p in train_root.iterdir() if p.is_dir()])
print("Classes:", classes)

# Reconstruct the documented 80/20 stratified training split.
# The official validation images remain the untouched test set.
items = []
labels = []

for class_name in classes:
    for image in sorted((train_root / class_name).glob("*")):
        if image.is_file():
            items.append(image)
            labels.append(class_name)

train_items, validation_items = train_test_split(
    items,
    test_size=1421,
    random_state=42,
    stratify=labels
)

# Build ImageFolder-compatible directories
DATA = Path("/content/cafr_data")
TEST = Path("/content/cafr_test")

def copy_items(files, destination):
    for source in files:
        class_name = source.parent.name
        target = destination / class_name / source.name
        target.parent.mkdir(parents=True, exist_ok=True)
        if not target.exists():
            os.symlink(source, target)

copy_items(train_items, DATA / "train")
copy_items(validation_items, DATA / "val")

# Official Imagenette validation set as held-out test set
for class_name in classes:
    source_dir = val_root / class_name
    target_dir = TEST / class_name
    target_dir.mkdir(parents=True, exist_ok=True)

    for source in source_dir.glob("*"):
        if source.is_file():
            target = target_dir / source.name
            if not target.exists():
                os.symlink(source, target)

print("\nDATASET PREPARATION")
print("Training images:", len(train_items))
print("Validation images:", len(validation_items))
print("Test images:", sum(1 for p in TEST.rglob("*") if p.is_file()))
print("Classes:", len(classes))
print("Training path:", DATA)
print("Test path:", TEST)
```

```python

import os
import sys
import json
import time
import shutil
import subprocess
from pathlib import Path
from datetime import datetime

import torch
import pandas as pd

# --------------------------------------------------
# 1. GPU AND GOOGLE DRIVE
# --------------------------------------------------
assert torch.cuda.is_available(), "CUDA GPU is unavailable."

from google.colab import drive
drive.mount("/content/drive")

print("GPU:", torch.cuda.get_device_name(0))

# --------------------------------------------------
# 2. PATHS AND EXPERIMENT CONFIGURATION
# --------------------------------------------------
REPO = Path("/content/cartoonization-robustness-study")
SOURCE = REPO / "research/CAFRNet_v2/source"

DRIVE_ROOT = Path("/content/drive/MyDrive/CAFRNet_v2_Experiments")
RUN_ROOT = DRIVE_ROOT / "imagenette_main_matrix"
LOG_ROOT = DRIVE_ROOT / "logs"

RUN_ROOT.mkdir(parents=True, exist_ok=True)
LOG_ROOT.mkdir(parents=True, exist_ok=True)

assert (SOURCE / "train_architecture.py").exists()
assert Path("/content/cafr_data/train").exists()
assert Path("/content/cafr_data/val").exists()

ARCHITECTURES = [
    "cafrnet_v2",
    "convnext_tiny",
    "resnet18",
]

SEEDS = [42, 123, 2026]

EPOCHS = 20
BATCH_SIZE = 16
WORKERS = 2
IMAGE_SIZE = 224

# --------------------------------------------------
# 3. SAVE EXPERIMENT CONFIGURATION
# --------------------------------------------------
config = {
    "dataset": "Imagenette 320",
    "train_images": 8048,
    "validation_images": 1421,
    "test_images": 3925,
    "classes": 10,
    "architectures": ARCHITECTURES,
    "seeds": SEEDS,
    "epochs": EPOCHS,
    "batch_size": BATCH_SIZE,
    "workers": WORKERS,
    "image_size": IMAGE_SIZE,
    "encoder_lr": 2e-5,
    "head_lr": 2e-4,
    "weight_decay": 0.05,
    "freeze_epochs": 2,
    "lambda_consistency": 0.25,
    "started_at": datetime.now().isoformat(),
}

with open(DRIVE_ROOT / "experiment_config.json", "w") as f:
    json.dump(config, f, indent=2)

# Copy the locked split manifest
manifest_source = (
    REPO / "research/CAFRNet_v2/manifests/"
    "imagenette_controlled_split.csv"
)
shutil.copy2(manifest_source, DRIVE_ROOT / manifest_source.name)

# --------------------------------------------------
# 4. RESUMABLE EXPERIMENT RUNNER
# --------------------------------------------------
results = []

for arch in ARCHITECTURES:
    for seed in SEEDS:

        run_name = f"{arch}_seed{seed}"
        run_dir = RUN_ROOT / run_name
        log_file = LOG_ROOT / f"{run_name}.log"
        done_file = run_dir / "COMPLETED.txt"

        run_dir.mkdir(parents=True, exist_ok=True)

        if done_file.exists():
            print(f"\nSKIPPING COMPLETED: {run_name}")
            results.append({
                "run": run_name,
                "status": "already_completed"
            })
            continue

        latest_checkpoint = run_dir / "latest_model.pt"

        command = [
            sys.executable, "-u",
            str(SOURCE / "train_architecture.py"),
            "--arch", arch,
            "--data", "/content/cafr_data",
            "--out", str(run_dir),
            "--epochs", str(EPOCHS),
            "--batch-size", str(BATCH_SIZE),
            "--seed", str(seed),
            "--workers", str(WORKERS),
            "--image-size", str(IMAGE_SIZE),
            "--encoder-lr", "2e-5",
            "--head-lr", "2e-4",
            "--weight-decay", "0.05",
            "--freeze-epochs", "2",
            "--lambda-consistency", "0.25",
        ]

        if latest_checkpoint.exists():
            command.append("--resume")
            print(f"\nRESUMING: {run_name}")
        else:
            print(f"\nSTARTING: {run_name}")

        print("Command:", " ".join(command))
        print("Output:", run_dir)

        start_time = time.time()
        environment = os.environ.copy()
        environment["PYTHONUNBUFFERED"] = "1"
        environment["OMP_NUM_THREADS"] = "2"

        return_code = None

        with open(log_file, "a", buffering=1) as log:
            log.write(
                f"\n\n===== {run_name} "
                f"{datetime.now().isoformat()} =====\n"
            )

            process = subprocess.Popen(
                command,
                cwd=str(SOURCE),
                stdout=subprocess.PIPE,
                stderr=subprocess.STDOUT,
                text=True,
                bufsize=1,
                env=environment
            )

            for line in process.stdout:
                print(line, end="")
                log.write(line)

            return_code = process.wait()

        elapsed = round((time.time() - start_time) / 60, 2)

        if return_code == 0 and (run_dir / "best_model.pt").exists():
            done_file.write_text(
                f"Completed: {datetime.now().isoformat()}\n"
                f"Elapsed minutes: {elapsed}\n"
            )
            status = "completed"
        else:
            status = "failed"

        results.append({
            "run": run_name,
            "status": status,
            "return_code": return_code,
            "elapsed_minutes": elapsed,
            "log": str(log_file)
        })

        pd.DataFrame(results).to_csv(
            DRIVE_ROOT / "matrix_progress.csv",
            index=False
        )

        print(f"\n{run_name}: {status}")
        print("Elapsed minutes:", elapsed)

        if status == "failed":
            print("Stopping matrix to protect remaining GPU time.")
            print("Inspect log:", log_file)
            break

    if results and results[-1].get("status") == "failed":
        break

# --------------------------------------------------
# 5. FINAL STATUS
# --------------------------------------------------
status_df = pd.DataFrame(results)
status_df.to_csv(DRIVE_ROOT / "matrix_progress.csv", index=False)

print("\n========== MATRIX STATUS ==========")
print(status_df.to_string(index=False))

print("\nResults folder:", RUN_ROOT)
print("Logs folder:", LOG_ROOT)
print("Progress file:", DRIVE_ROOT / "matrix_progress.csv")

print("\nGPU:", torch.cuda.get_device_name(0))
print("GPU memory allocated:",
      round(torch.cuda.memory_allocated() / 1024**3, 2), "GB")
print("GPU memory reserved:",
      round(torch.cuda.memory_reserved() / 1024**3, 2), "GB")

# ==================================================
# 6. CREATE DOWNLOADABLE ZIP BACKUP
# ==================================================
import zipfile
from datetime import datetime
from pathlib import Path

ZIP_DIR = Path("/content/drive/MyDrive/CAFRNet_v2_Experiments")
ZIP_DIR.mkdir(parents=True, exist_ok=True)

ZIP_PATH = ZIP_DIR / "CAFRNet_v2_Imagenette_Main_Matrix_Backup.zip"

print("\nCreating experiment backup ZIP...")

with zipfile.ZipFile(
    ZIP_PATH, "w", compression=zipfile.ZIP_DEFLATED,
    compresslevel=3
) as archive:

    # All experiment checkpoints, metadata and completion markers
    for file in RUN_ROOT.rglob("*"):
        if file.is_file():
            archive.write(
                file,
                arcname=f"imagenette_main_matrix/{file.relative_to(RUN_ROOT)}"
            )

    # Training logs
    for file in LOG_ROOT.rglob("*"):
        if file.is_file():
            archive.write(
                file,
                arcname=f"logs/{file.relative_to(LOG_ROOT)}"
            )

    # Configuration and split manifest
    for name in [
        "experiment_config.json",
        "imagenette_controlled_split.csv",
        "matrix_progress.csv"
    ]:
        file = ZIP_DIR / name
        if file.exists():
            archive.write(file, arcname=f"metadata/{name}")

size_mb = ZIP_PATH.stat().st_size / (1024 ** 2)

print("\nBACKUP ZIP CREATED")
print("Location:", ZIP_PATH)
print("Size:", round(size_mb, 2), "MB")

# Copy a second ZIP into the Colab session for direct download
LOCAL_ZIP = Path("/content/CAFRNet_v2_Imagenette_Main_Matrix_Backup.zip")
shutil.copy2(ZIP_PATH, LOCAL_ZIP)

print("Local download copy:", LOCAL_ZIP)

# Automatically download to your computer
from google.colab import files
files.download(str(LOCAL_ZIP))
```

```python

import os
import sys
import time
import shutil
import subprocess
from pathlib import Path

import torch

REPO = Path("/content/cartoonization-robustness-study")
SOURCE = REPO / "research/CAFRNet_v2/source"
SCRIPT = SOURCE / "train_architecture.py"

DRIVE_ROOT = Path("/content/drive/MyDrive/CAFRNet_v2_Experiments")
DRIVE_RUN = DRIVE_ROOT / "imagenette_main_matrix/cafrnet_v2_seed42"
LOG_FILE = DRIVE_ROOT / "logs/cafrnet_v2_seed42.log"

DRIVE_RUN.mkdir(parents=True, exist_ok=True)
LOG_FILE.parent.mkdir(parents=True, exist_ok=True)

# 1. Inspect the saved checkpoint without changing it
checkpoint_path = DRIVE_RUN / "latest_model.pt"
assert checkpoint_path.exists(), "Resume checkpoint not found in Drive."

checkpoint = torch.load(
    checkpoint_path,
    map_location="cpu",
    weights_only=False
)

state = checkpoint.get("loader_generator_state")
print("Checkpoint epoch:", checkpoint.get("epoch"))
print("Generator state type:", type(state))
if isinstance(state, torch.Tensor):
    print("Generator state dtype:", state.dtype)
    print("Generator state shape:", tuple(state.shape))

# 2. Patch generator state conversion for PyTorch compatibility
source = SCRIPT.read_text()

old = 'generator.set_state(ck["loader_generator_state"])'
new = '''generator_state = ck["loader_generator_state"]
            if not isinstance(generator_state, torch.Tensor):
                generator_state = torch.as_tensor(generator_state, dtype=torch.uint8)
            else:
                generator_state = generator_state.detach().cpu().to(torch.uint8)
            generator.set_state(generator_state)'''

if old in source:
    source = source.replace(old, new, 1)
    SCRIPT.write_text(source)
    print("Resume compatibility patch applied.")
elif new in source:
    print("Compatibility patch already present.")
else:
    raise RuntimeError("Could not locate the expected generator-state line.")

# 3. Compile patched script
subprocess.run(
    [sys.executable, "-m", "py_compile", str(SCRIPT)],
    check=True
)
print("Training script syntax: OK")

# 4. Train directly into Google Drive.
# This means latest_model.pt is saved persistently after each epoch.
command = [
    sys.executable, "-u", str(SCRIPT),
    "--arch", "cafrnet_v2",
    "--data", "/content/cafr_data",
    "--out", str(DRIVE_RUN),
    "--epochs", "20",
    "--batch-size", "16",
    "--seed", "42",
    "--workers", "2",
    "--image-size", "224",
    "--encoder-lr", "2e-5",
    "--head-lr", "2e-4",
    "--weight-decay", "0.05",
    "--freeze-epochs", "2",
    "--lambda-consistency", "0.25",
    "--resume"
]

print("\nResuming CAFR-Net v2 — seed 42")
print("Saved epoch:", checkpoint.get("epoch"))
print("Persistent output:", DRIVE_RUN)

env = os.environ.copy()
env["PYTHONUNBUFFERED"] = "1"
env["OMP_NUM_THREADS"] = "2"

start = time.time()

with open(LOG_FILE, "a", buffering=1) as log:
    process = subprocess.Popen(
        command,
        cwd=str(SOURCE),
        stdout=subprocess.PIPE,
        stderr=subprocess.STDOUT,
        text=True,
        bufsize=1,
        env=env
    )

    for line in process.stdout:
        print(line, end="")
        log.write(line)

    exit_code = process.wait()

print("\nExit code:", exit_code)
print("Elapsed minutes:", round((time.time() - start) / 60, 2))

# 5. Verify persistent progress
if checkpoint_path.exists():
    latest = torch.load(
        checkpoint_path,
        map_location="cpu",
        weights_only=False
    )
    print("Latest saved epoch:", latest.get("epoch"))

print("Best checkpoint exists:", (DRIVE_RUN / "best_model.pt").exists())
print("Latest checkpoint exists:", checkpoint_path.exists())

if exit_code != 0:
    print("Training stopped. The latest Drive checkpoint is preserved.")
else:
    print("Training process completed.")
```

```python

from pathlib import Path
import json
import pandas as pd
import numpy as np

REPO = Path("/content/cartoonization-robustness-study")
BASE = REPO / "research/CAFRNet_v2/results/imagenette"

runs = [
    ("CAFRNet", "seed42"),
    ("CAFRNet", "seed123"),
    ("CAFRNet", "seed2026"),
    ("No edge", "ablation_no_edge_seed42"),
    ("No semantic fusion", "ablation_no_semantic_fusion_seed42"),
    ("No attention", "ablation_no_attention_seed42"),
]

audit = []
prediction_frames = []

for model, folder in runs:
    run_dir = BASE / folder
    metadata_file = run_dir / "run_metadata.json"
    history_file = run_dir / "history.json"
    prediction_file = run_dir / "test_predictions.csv"

    row = {
        "model": model,
        "run": folder,
        "metadata_exists": metadata_file.exists(),
        "history_exists": history_file.exists(),
        "predictions_exist": prediction_file.exists()
    }

    if metadata_file.exists():
        metadata = json.loads(metadata_file.read_text())
        args = metadata.get("args", {})
        row.update({
            "seed": metadata.get("seed", args.get("seed")),
            "epochs_recorded": args.get("epochs"),
            "batch_size": args.get("batch_size"),
            "gpu": metadata.get("gpu"),
            "train_samples": metadata.get("train_samples"),
            "validation_samples": metadata.get("validation_samples")
        })

    if history_file.exists():
        history = json.loads(history_file.read_text())
        row["history_epochs"] = len(history)
        row["last_epoch"] = history[-1].get("epoch") if history else None
        row["best_val_accuracy"] = (
            max(x.get("val_accuracy", 0) for x in history)
            if history else None
        )

    if prediction_file.exists():
        df = pd.read_csv(prediction_file)
        row["prediction_rows"] = len(df)
        row["prediction_columns"] = ",".join(df.columns)

        if "condition" in df.columns:
            row["conditions"] = ",".join(sorted(df.condition.astype(str).unique()))

        df["run"] = folder
        df["model_group"] = model
        prediction_frames.append(df)

    audit.append(row)

audit_df = pd.DataFrame(audit)

print("===== EXISTING EXPERIMENT AUDIT =====")
print(audit_df.to_string(index=False))

if prediction_frames:
    predictions = pd.concat(prediction_frames, ignore_index=True)
    print("\n===== PREDICTION DATA SUMMARY =====")
    print("Total prediction rows:", len(predictions))
    print("Available columns:", predictions.columns.tolist())

    if "condition" in predictions.columns:
        print("\nRows by run and condition:")
        print(
            predictions.groupby(["run", "condition"])
            .size().to_string()
        )

    predictions.to_csv(
        "/content/CAFRNet_existing_predictions_audit.csv",
        index=False
    )

# Save audit
audit_df.to_csv(
    "/content/CAFRNet_existing_experiment_audit.csv",
    index=False
)

print("\nAudit files saved:")
print("/content/CAFRNet_existing_experiment_audit.csv")
print("/content/CAFRNet_existing_predictions_audit.csv")
```

```python

# STEP 13: Statistical analysis of existing CAFR-Net experiments
# CPU only — NO TRAINING, NO GPU REQUIRED

from pathlib import Path
import pandas as pd
import numpy as np
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score, f1_score
)
from scipy.stats import binomtest
from statsmodels.stats.multitest import multipletests

BASE = Path("/content/cartoonization-robustness-study/research/CAFRNet_v2/results/imagenette")
INPUT = Path("/content/CAFRNet_existing_predictions_audit.csv")
OUT = BASE / "final_statistical_audit"
OUT.mkdir(parents=True, exist_ok=True)

df = pd.read_csv(INPUT)
df["correct"] = df["correct"].astype(str).str.lower().isin(["true", "1"])
df["true_label"] = df["true_label"].astype(str)
df["predicted_label"] = df["predicted_label"].astype(str)

conditions = ["original", "opencv", "animegan2", "stable_diffusion"]
full_runs = ["seed42", "seed123", "seed2026"]
ablations = {
    "No edge": "ablation_no_edge_seed42",
    "No semantic fusion": "ablation_no_semantic_fusion_seed42",
    "No attention": "ablation_no_attention_seed42"
}

rng = np.random.default_rng(42)
N_BOOT = 2000

def metric_values(y, pred):
    return {
        "accuracy": accuracy_score(y, pred),
        "macro_precision": precision_score(y, pred, average="macro", zero_division=0),
        "macro_recall": recall_score(y, pred, average="macro", zero_division=0),
        "macro_f1": f1_score(y, pred, average="macro", zero_division=0)
    }

def bootstrap_metric_ci(y, pred, n_boot=N_BOOT):
    y = np.asarray(y)
    pred = np.asarray(pred)
    n = len(y)
    values = {k: [] for k in ["accuracy", "macro_precision", "macro_recall", "macro_f1"]}

    for _ in range(n_boot):
        idx = rng.integers(0, n, n)
        m = metric_values(y[idx], pred[idx])
        for k in values:
            values[k].append(m[k])

    return {
        k: (np.percentile(v, 2.5), np.percentile(v, 97.5))
        for k, v in values.items()
    }

# 1. Validate test-set identity and label consistency
identity_checks = []
for condition in conditions:
    subset = df[df.condition == condition]
    reference = None

    for run in sorted(subset.run.unique()):
        current = subset[subset.run == run].sort_values("relative_path")
        keys = current[["relative_path", "true_label"]].reset_index(drop=True)

        if reference is None:
            reference = keys
            matches = True
        else:
            matches = keys.equals(reference)

        identity_checks.append({
            "condition": condition,
            "run": run,
            "same_test_images_and_labels": matches,
            "number_of_images": len(current),
            "unique_image_paths": current.relative_path.nunique()
        })

identity_df = pd.DataFrame(identity_checks)
identity_df.to_csv(OUT / "test_identity_validation.csv", index=False)

if not identity_df["same_test_images_and_labels"].all():
    raise ValueError(
        "Test identity mismatch detected. Stop before paired comparisons."
    )

# 2. Per-run metrics and image-bootstrap confidence intervals
metric_rows = []

for run in sorted(df.run.unique()):
    for condition in conditions:
        part = df[(df.run == run) & (df.condition == condition)].sort_values("relative_path")
        y = part.true_label.to_numpy()
        pred = part.predicted_label.to_numpy()

        metrics = metric_values(y, pred)
        cis = bootstrap_metric_ci(y, pred)

        row = {"run": run, "condition": condition, "n": len(part)}
        for metric, value in metrics.items():
            row[metric] = value * 100
            row[metric + "_ci95_low"] = cis[metric][0] * 100
            row[metric + "_ci95_high"] = cis[metric][1] * 100

        metric_rows.append(row)

metrics_df = pd.DataFrame(metric_rows)
metrics_df.to_csv(OUT / "per_run_metrics_with_bootstrap_ci.csv", index=False)

# 3. Three-seed CAFR-Net summary: mean, SD and range across seeds
seed_df = metrics_df[metrics_df.run.isin(full_runs)]
seed_summary = (
    seed_df.groupby("condition")
    .agg(
        seeds=("run", "nunique"),
        accuracy_mean=("accuracy", "mean"),
        accuracy_sd=("accuracy", "std"),
        accuracy_min=("accuracy", "min"),
        accuracy_max=("accuracy", "max"),
        macro_f1_mean=("macro_f1", "mean"),
        macro_f1_sd=("macro_f1", "std")
    )
    .reset_index()
)
seed_summary.to_csv(OUT / "three_seed_cafrnet_summary.csv", index=False)

# 4. Paired ablation comparison against full CAFR-Net seed 42
paired_rows = []

for condition in conditions:
    base = df[
        (df.run == "seed42") & (df.condition == condition)
    ].sort_values("relative_path")

    y = base.true_label.to_numpy()
    full_pred = base.predicted_label.to_numpy()
    full_correct = (full_pred == y)

    for ablation_name, ablation_run in ablations.items():
        alt = df[
            (df.run == ablation_run) & (df.condition == condition)
        ].sort_values("relative_path")

        if not np.array_equal(base.relative_path.to_numpy(),
                              alt.relative_path.to_numpy()):
            raise ValueError(f"Image order mismatch: {ablation_name}, {condition}")

        alt_pred = alt.predicted_label.to_numpy()
        alt_correct = (alt_pred == y)

        # Positive delta means full CAFR-Net is better
        delta = full_correct.astype(int) - alt_correct.astype(int)
        observed = delta.mean() * 100

        # Paired bootstrap over identical test-image positions
        boot = np.empty(N_BOOT)
        n = len(delta)
        for b in range(N_BOOT):
            idx = rng.integers(0, n, n)
            boot[b] = delta[idx].mean() * 100

        # Exact two-sided McNemar test
        full_only = int(np.sum(full_correct & ~alt_correct))
        ablation_only = int(np.sum(~full_correct & alt_correct))
        discordant = full_only + ablation_only

        p_value = (
            binomtest(
                min(full_only, ablation_only),
                discordant,
                0.5,
                alternative="two-sided"
            ).pvalue if discordant else 1.0
        )

        paired_rows.append({
            "condition": condition,
            "ablation": ablation_name,
            "n": n,
            "full_accuracy_pct": full_correct.mean() * 100,
            "ablation_accuracy_pct": alt_correct.mean() * 100,
            "full_minus_ablation_pp": observed,
            "paired_bootstrap_ci95_low_pp": np.percentile(boot, 2.5),
            "paired_bootstrap_ci95_high_pp": np.percentile(boot, 97.5),
            "full_only_correct": full_only,
            "ablation_only_correct": ablation_only,
            "discordant_pairs": discordant,
            "mcnemar_p_raw": p_value
        })

paired_df = pd.DataFrame(paired_rows)
paired_df["mcnemar_p_holm"] = multipletests(
    paired_df["mcnemar_p_raw"], method="holm"
)[1]
paired_df.to_csv(OUT / "paired_ablation_statistics.csv", index=False)

# 5. Print results
print("\n===== TEST IDENTITY VALIDATION =====")
print(identity_df.to_string(index=False))

print("\n===== THREE-SEED CAFR-NET SUMMARY (%) =====")
print(seed_summary.round(4).to_string(index=False))

print("\n===== PER-RUN METRICS (%) =====")
print(metrics_df.round(4).to_string(index=False))

print("\n===== PAIRED ABLATION ANALYSIS =====")
print(paired_df.round(5).to_string(index=False))

print("\nAll analysis files saved in:")
print(OUT)

print("\nFiles:")
for p in sorted(OUT.glob("*.csv")):
    print("-", p.name)
```

```python

# STEP 13 — OPTIMIZED CAFR-Net STATISTICAL ANALYSIS
# CPU ONLY | Reuses existing predictions | No model training

from pathlib import Path
import pandas as pd
import numpy as np
from scipy.stats import binomtest
from statsmodels.stats.multitest import multipletests
from sklearn.metrics import precision_recall_fscore_support
import time

START = time.time()
INPUT = Path("/content/CAFRNet_existing_predictions_audit.csv")
OUT = Path(
    "/content/cartoonization-robustness-study/research/"
    "CAFRNet_v2/results/imagenette/final_statistical_audit"
)
OUT.mkdir(parents=True, exist_ok=True)

N_BOOT = 500
SEED = 42
rng = np.random.default_rng(SEED)

conditions = ["original", "opencv", "animegan2", "stable_diffusion"]
full_runs = ["seed42", "seed123", "seed2026"]

ablations = {
    "No edge": "ablation_no_edge_seed42",
    "No semantic fusion": "ablation_no_semantic_fusion_seed42",
    "No attention": "ablation_no_attention_seed42"
}

print("Loading existing predictions...", flush=True)
df = pd.read_csv(INPUT)

df["correct"] = (
    df["correct"].astype(str).str.lower().isin(["true", "1"])
)
df["true_label"] = df["true_label"].astype(str)
df["predicted_label"] = df["predicted_label"].astype(str)

# -----------------------------------------------
# 1. TEST IDENTITY VALIDATION
# -----------------------------------------------

print("Validating test image identities...", flush=True)

identity_rows = []

for condition in conditions:
    subset = df[df.condition == condition]
    reference = None

    for run in sorted(subset.run.unique()):
        part = subset[subset.run == run].sort_values(
            "relative_path"
        )

        keys = part[
            ["relative_path", "true_label"]
        ].reset_index(drop=True)

        same = True if reference is None else keys.equals(reference)

        if reference is None:
            reference = keys

        identity_rows.append({
            "condition": condition,
            "run": run,
            "same_test_images_and_labels": same,
            "images": len(part),
            "unique_paths": part.relative_path.nunique()
        })

identity_df = pd.DataFrame(identity_rows)
identity_df.to_csv(
    OUT / "test_identity_validation.csv", index=False
)

if not identity_df.same_test_images_and_labels.all():
    raise ValueError(
        "Test identity mismatch. Paired analysis stopped."
    )

# -----------------------------------------------
# 2. FAST VECTORIZED METRICS + BOOTSTRAP
# -----------------------------------------------

print("Calculating per-run metrics and bootstrap CIs...", flush=True)

metric_rows = []

def calculate_metrics(y, pred, classes):
    precision, recall, f1, _ = precision_recall_fscore_support(
        y, pred,
        labels=classes,
        average="macro",
        zero_division=0
    )
    accuracy = np.mean(y == pred)

    return np.array([
        accuracy, precision, recall, f1
    ])

for run in sorted(df.run.unique()):
    for condition in conditions:
        part = df[
            (df.run == run) &
            (df.condition == condition)
        ].sort_values("relative_path")

        y = part.true_label.to_numpy()
        pred = part.predicted_label.to_numpy()
        classes = np.unique(y)

        observed = calculate_metrics(y, pred, classes)

        # Vectorized bootstrap for accuracy.
        # Macro metrics are calculated on the original test set.
        # Accuracy CI uses paired image-level resampling.
        correct = (y == pred).astype(np.float64)
        n = len(correct)

        boot_acc = np.empty(N_BOOT)

        for b in range(N_BOOT):
            idx = rng.integers(0, n, size=n)
            boot_acc[b] = correct[idx].mean() * 100

        row = {
            "run": run,
            "condition": condition,
            "n": n,
            "accuracy_pct": observed[0] * 100,
            "accuracy_ci95_low": np.percentile(boot_acc, 2.5),
            "accuracy_ci95_high": np.percentile(boot_acc, 97.5),
            "macro_precision_pct": observed[1] * 100,
            "macro_recall_pct": observed[2] * 100,
            "macro_f1_pct": observed[3] * 100
        }

        metric_rows.append(row)

metrics_df = pd.DataFrame(metric_rows)
metrics_df.to_csv(
    OUT / "per_run_metrics_with_bootstrap_ci.csv",
    index=False
)

# -----------------------------------------------
# 3. THREE-SEED SUMMARY
# -----------------------------------------------

print("Preparing three-seed summary...", flush=True)

seed_df = metrics_df[
    metrics_df.run.isin(full_runs)
]

seed_summary = (
    seed_df.groupby("condition")
    .agg(
        seeds=("run", "nunique"),
        accuracy_mean=("accuracy_pct", "mean"),
        accuracy_sd=("accuracy_pct", "std"),
        accuracy_min=("accuracy_pct", "min"),
        accuracy_max=("accuracy_pct", "max"),
        macro_f1_mean=("macro_f1_pct", "mean"),
        macro_f1_sd=("macro_f1_pct", "std")
    )
    .reset_index()
)

seed_summary.to_csv(
    OUT / "three_seed_cafrnet_summary.csv",
    index=False
)

# -----------------------------------------------
# 4. PAIRED ABLATION ANALYSIS
# -----------------------------------------------

print("Running paired ablation comparisons...", flush=True)

paired_rows = []

for condition in conditions:
    base = df[
        (df.run == "seed42") &
        (df.condition == condition)
    ].sort_values("relative_path")

    y = base.true_label.to_numpy()
    full_pred = base.predicted_label.to_numpy()
    full_correct = (full_pred == y)

    for ablation_name, ablation_run in ablations.items():

        alt = df[
            (df.run == ablation_run) &
            (df.condition == condition)
        ].sort_values("relative_path")

        if not np.array_equal(
            base.relative_path.to_numpy(),
            alt.relative_path.to_numpy()
        ):
            raise ValueError(
                f"Image mismatch: {ablation_name}, {condition}"
            )

        alt_pred = alt.predicted_label.to_numpy()
        alt_correct = (alt_pred == y)

        delta = (
            full_correct.astype(np.int8) -
            alt_correct.astype(np.int8)
        )

        observed = delta.mean() * 100

        # Paired bootstrap: resample the same image-level differences.
        n = len(delta)
        boot = np.empty(N_BOOT)

        for b in range(N_BOOT):
            idx = rng.integers(0, n, size=n)
            boot[b] = delta[idx].mean() * 100

        full_only = int(np.sum(full_correct & ~alt_correct))
        alt_only = int(np.sum(~full_correct & alt_correct))
        discordant = full_only + alt_only

        p = (
            binomtest(
                min(full_only, alt_only),
                discordant,
                0.5,
                alternative="two-sided"
            ).pvalue
            if discordant else 1.0
        )

        paired_rows.append({
            "condition": condition,
            "ablation": ablation_name,
            "n": n,
            "full_accuracy_pct": full_correct.mean() * 100,
            "ablation_accuracy_pct": alt_correct.mean() * 100,
            "full_minus_ablation_pp": observed,
            "paired_bootstrap_ci95_low_pp": np.percentile(boot, 2.5),
            "paired_bootstrap_ci95_high_pp": np.percentile(boot, 97.5),
            "full_only_correct": full_only,
            "ablation_only_correct": alt_only,
            "discordant_pairs": discordant,
            "mcnemar_p_raw": p
        })

paired_df = pd.DataFrame(paired_rows)

paired_df["mcnemar_p_holm"] = multipletests(
    paired_df["mcnemar_p_raw"],
    method="holm"
)[1]

paired_df.to_csv(
    OUT / "paired_ablation_statistics.csv",
    index=False
)

# -----------------------------------------------
# 5. SAVE AND DISPLAY RESULTS
# -----------------------------------------------

print("\n===== TEST IDENTITY VALIDATION =====")
print(identity_df.to_string(index=False))

print("\n===== THREE-SEED CAFR-NET SUMMARY (%) =====")
print(seed_summary.round(4).to_string(index=False))

print("\n===== PER-RUN METRICS (%) =====")
print(metrics_df.round(4).to_string(index=False))

print("\n===== PAIRED ABLATION RESULTS =====")
print(paired_df.round(5).to_string(index=False))

print("\n===== OUTPUT FILES =====")
for file in sorted(OUT.glob("*.csv")):
    print(file.name)

print("\nTotal time: %.1f seconds" % (time.time() - START))
print("STATISTICAL ANALYSIS COMPLETED")
```

```python
from pathlib import Path
from torchvision.datasets import ImageFolder
import torch
import os, shutil, subprocess, sys

root = Path("/content/cartoonization-robustness-study")
original = Path("/content/cafr_test")
degraded = root / "results/degraded"

checkpoint = Path(
    "/content/cafr_runs/controlled_imagenette/cafrnet_seed42/best_model.pt"
)

# Load original test labels and checkpoint class order
base = ImageFolder(str(original))
ck = torch.load(checkpoint, map_location="cpu", weights_only=False)
classes = list(ck["classes"])

if base.classes != classes:
    raise ValueError(
        f"Class order mismatch!\nDataset: {base.classes}\nCheckpoint: {classes}"
    )

# Index original test images by filename stem
lookup = {}
ambiguous = set()

for path, label in base.samples:
    stem = Path(path).stem.lower()
    if stem in lookup:
        ambiguous.add(stem)
    else:
        lookup[stem] = (classes[label], path)

for stem in ambiguous:
    lookup.pop(stem, None)

# Prepare temporary ImageFolder structures
workspace = Path("/content/cafr_runs/seed42_stress_imagefolders")
if workspace.exists():
    shutil.rmtree(workspace)
workspace.mkdir(parents=True)

conditions = []
report = []

for folder in sorted(degraded.iterdir()):
    if not folder.is_dir():
        continue

    destination = workspace / folder.name
    for cls in classes:
        (destination / cls).mkdir(parents=True, exist_ok=True)

    matched = 0
    unmatched = 0

    for image_path in folder.rglob("*"):
        if not image_path.is_file() or image_path.suffix.lower() not in [
            ".jpg", ".jpeg", ".png", ".webp", ".bmp"
        ]:
            continue

        stem = image_path.stem.lower()
        item = lookup.get(stem)

        if item is None:
            unmatched += 1
            continue

        cls, _ = item
        target = destination / cls / f"{matched:05d}_{image_path.name}"

        os.symlink(image_path.resolve(), target)
        matched += 1

    if matched:
        conditions.append(f"{folder.name}={destination}")
        report.append((folder.name, matched, unmatched))

print("=== DATASET MAPPING ===")
for name, matched, unmatched in report:
    print(f"{name:40} matched={matched:4} unmatched={unmatched:4}")

if not conditions:
    raise RuntimeError("No images matched the original test dataset.")

print("\nEvaluating", len(conditions), "conditions")
print("Total mapped images:", sum(x[1] for x in report))

# Run corrected seed-42 evaluation
source = root / "research/CAFRNet_v2/source"
output = Path("/content/cafr_runs/seed42_stress_predictions.csv")

cmd = [
    sys.executable, "-u", str(source / "evaluate_architecture.py"),
    "--checkpoint", str(checkpoint),
    "--conditions", *conditions,
    "--out", str(output),
    "--batch-size", "32",
    "--workers", "2",
    "--image-size", "224"
]

subprocess.run(cmd, check=True)

print("\nSUCCESS!")
print("Stress predictions:", output)
print("Rows:", sum(x[1] for x in report))
```

```python
from pathlib import Path
import pandas as pd
import numpy as np
import json, zipfile, re
import matplotlib.pyplot as plt
from scipy.stats import binomtest
from sklearn.metrics import f1_score

base = Path("/content/cafr_runs/evaluation_seed42")
stress_path = Path("/content/cafr_runs/seed42_stress_predictions.csv")
root = Path("/content/cartoonization-robustness-study")
out = root / "research/CAFRNet_v2/results/seed42_corrected_robustness"
out.mkdir(parents=True, exist_ok=True)

original = pd.read_csv(base)
stress = pd.read_csv(stress_path)

# Recover the original filename from the temporary ImageFolder symlink name.
def image_key(row, stress_row=False):
    rel = str(row["relative_path"]).replace("\\", "/")
    parts = rel.split("/")
    filename = parts[-1]
    if stress_row:
        filename = re.sub(r"^\d+_", "", filename)
    return str(row["true_label"]) + "/" + filename.lower()

original["match_key"] = original.apply(
    lambda r: image_key(r), axis=1
)
stress["match_key"] = stress.apply(
    lambda r: image_key(r, True), axis=1
)

if original["match_key"].duplicated().any():
    raise ValueError("Duplicate original test identities detected.")

if stress.duplicated(["condition", "match_key"]).any():
    raise ValueError("Duplicate stress identities detected.")

original_subset = original.set_index("match_key")

summary_rows = []
paired_rows = []
class_rows = []

rng = np.random.default_rng(42)

for condition, group in stress.groupby("condition"):
    group = group.copy()
    group["original_correct"] = group["match_key"].map(
        original_subset["correct"]
    )

    if group["original_correct"].isna().any():
        raise ValueError(f"Unmatched original identities in {condition}")

    y = group["true_label"].astype(str)
    pred = group["predicted_label"].astype(str)

    stress_correct = group["correct"].astype(int).to_numpy()
    original_correct = group["original_correct"].astype(int).to_numpy()

    n = len(group)
    delta = stress_correct - original_correct

    # Paired bootstrap CI for accuracy difference
    boot = np.empty(5000)
    for i in range(len(boot)):
        idx = rng.integers(0, n, n)
        boot[i] = delta[idx].mean() * 100

    ci_low, ci_high = np.percentile(boot, [2.5, 97.5])

    # Exact paired McNemar test
    b = int(((original_correct == 1) & (stress_correct == 0)).sum())
    c = int(((original_correct == 0) & (stress_correct == 1)).sum())
    discordant = b + c
    p = binomtest(min(b, c), discordant, 0.5).pvalue if discordant else 1.0

    original_acc = original_correct.mean()
    stress_acc = stress_correct.mean()

    summary_rows.append({
        "condition": condition,
        "n": n,
        "original_subset_accuracy": original_acc,
        "stress_accuracy": stress_acc,
        "accuracy_change_pp": (stress_acc-original_acc)*100,
        "delta_ci95_low_pp": ci_low,
        "delta_ci95_high_pp": ci_high,
        "stress_macro_f1": f1_score(
            y, pred, average="macro", zero_division=0
        ),
        "mean_confidence": group["confidence"].mean()
    })

    paired_rows.append({
        "condition": condition,
        "n": n,
        "original_correct_stress_wrong": b,
        "original_wrong_stress_correct": c,
        "discordant_pairs": discordant,
        "mcnemar_exact_p": p
    })

    for cls, cg in group.groupby("true_label"):
        class_rows.append({
            "condition": condition,
            "class": cls,
            "n": len(cg),
            "accuracy": cg["correct"].mean()
        })

summary = pd.DataFrame(summary_rows)
paired = pd.DataFrame(paired_rows)
per_class = pd.DataFrame(class_rows)

# Holm correction across 15 paired tests
pvals = paired["mcnemar_exact_p"].to_numpy()
order = np.argsort(pvals)
adjusted = np.empty(len(pvals))
running = 0
m = len(pvals)

for rank, idx in enumerate(order):
    value = min(1.0, (m-rank)*pvals[idx])
    running = max(running, value)
    adjusted[idx] = running

paired["mcnemar_holm_p"] = adjusted

summary.to_csv(out / "seed42_robustness_summary.csv", index=False)
paired.to_csv(out / "seed42_paired_mcnemar.csv", index=False)
per_class.to_csv(out / "seed42_per_class_metrics.csv", index=False)

# Plot accuracy by condition
plot_df = summary.sort_values("stress_accuracy")

plt.figure(figsize=(11, 7))
plt.barh(plot_df["condition"], plot_df["stress_accuracy"] * 100)
plt.xlabel("Accuracy (%)")
plt.title("Corrected CAFR-Net v2 — Seed 42 Robustness")
plt.xlim(0, 105)
plt.tight_layout()
plt.savefig(out / "seed42_robustness_accuracy.png", dpi=200)
plt.close()

metadata = {
    "architecture": "CAFR-Net v2",
    "seed": 42,
    "training_epochs": 20,
    "best_checkpoint_epoch": 2,
    "best_validation_accuracy": 0.9985925404644617,
    "original_test_size": len(original),
    "stress_prediction_rows": len(stress),
    "conditions": int(summary.shape[0]),
    "sample_size_per_condition": sorted(
        summary["n"].unique().tolist()
    ),
    "paired_by": "true class and original image filename",
    "statistics": [
        "paired bootstrap 95% confidence interval",
        "exact McNemar test",
        "Holm correction across conditions"
    ],
    "limitation": "Stress conditions use 300-image samples per condition."
}
(out / "experiment_metadata.json").write_text(
    json.dumps(metadata, indent=2)
)

# Include raw predictions and analysis in one archive
with zipfile.ZipFile(
    out / "CAFRNet_seed42_robustness_results.zip", "w",
    zipfile.ZIP_DEFLATED
) as z:
    for f in out.iterdir():
        if f.is_file() and f.name != "CAFRNet_seed42_robustness_results.zip":
            z.write(f, arcname=f.name)
    z.write(stress_path, arcname="raw/seed42_stress_predictions.csv")
    z.write(base, arcname="raw/seed42_original_predictions.csv")

print("PAIRED ROBUSTNESS ANALYSIS COMPLETE")
display(summary.round(4))
print("\nPAIRED SIGNIFICANCE TESTS")
display(paired.round(6))
print("\nRESULT PACKAGE:")
print(out / "CAFRNet_seed42_robustness_results.zip")
```

```python
from pathlib import Path
import pandas as pd
import numpy as np
import json, zipfile
import matplotlib.pyplot as plt
from scipy.stats import binomtest
from sklearn.metrics import f1_score

base_path = Path("/content/cafr_runs/evaluation_seed42")
stress_path = Path("/content/cafr_runs/seed42_stress_predictions.csv")
root = Path("/content/cartoonization-robustness-study")
out = root / "research/CAFRNet_v2/results/seed42_corrected_robustness"
out.mkdir(parents=True, exist_ok=True)

original = pd.read_csv(base_path)
stress = pd.read_csv(stress_path)

# Match class + filename stem, ignoring .jpg / .JPEG differences
original["match_key"] = original.apply(
    lambda r: str(r["true_label"]) + "/" +
    Path(str(r["relative_path"]).replace("\\", "/")).stem.lower(),
    axis=1
)

stress["match_key"] = stress.apply(
    lambda r: str(r["true_label"]) + "/" +
    Path(
        str(r["relative_path"]).replace("\\", "/").split("/")[-1]
        .removeprefix("00000_")
    ).stem.lower(),
    axis=1
)

# Remove the numeric link prefix robustly, regardless of its length
import re
stress["match_key"] = stress.apply(
    lambda r: str(r["true_label"]) + "/" +
    Path(re.sub(
        r"^\d+_", "",
        str(r["relative_path"]).replace("\\", "/").split("/")[-1]
    )).stem.lower(),
    axis=1
)

if original["match_key"].duplicated().any():
    raise ValueError("Duplicate original image identities detected.")

if stress.duplicated(["condition", "match_key"]).any():
    raise ValueError("Duplicate stress image identities detected.")

original_lookup = original.set_index("match_key")["correct"]

summary_rows, paired_rows, class_rows = [], [], []
rng = np.random.default_rng(42)

for condition, group in stress.groupby("condition"):
    group = group.copy()
    group["original_correct"] = group["match_key"].map(original_lookup)

    unmatched = group["original_correct"].isna().sum()
    if unmatched:
        examples = group.loc[
            group["original_correct"].isna(), "match_key"
        ].head().tolist()
        raise ValueError(
            f"{condition}: {unmatched} unmatched images. Examples: {examples}"
        )

    y = group["true_label"].astype(str)
    pred = group["predicted_label"].astype(str)

    stress_correct = group["correct"].astype(int).to_numpy()
    original_correct = group["original_correct"].astype(int).to_numpy()

    n = len(group)
    delta = stress_correct - original_correct

    boot = np.array([
        delta[rng.integers(0, n, n)].mean() * 100
        for _ in range(5000)
    ])
    ci_low, ci_high = np.percentile(boot, [2.5, 97.5])

    b = int(((original_correct == 1) & (stress_correct == 0)).sum())
    c = int(((original_correct == 0) & (stress_correct == 1)).sum())
    discordant = b + c
    p = binomtest(min(b, c), discordant, 0.5).pvalue if discordant else 1.0

    summary_rows.append({
        "condition": condition,
        "n": n,
        "original_subset_accuracy": original_correct.mean(),
        "stress_accuracy": stress_correct.mean(),
        "accuracy_change_pp": delta.mean() * 100,
        "delta_ci95_low_pp": ci_low,
        "delta_ci95_high_pp": ci_high,
        "stress_macro_f1": f1_score(y, pred, average="macro", zero_division=0),
        "mean_confidence": group["confidence"].mean()
    })

    paired_rows.append({
        "condition": condition,
        "n": n,
        "original_correct_stress_wrong": b,
        "original_wrong_stress_correct": c,
        "discordant_pairs": discordant,
        "mcnemar_exact_p": p
    })

    for cls, cg in group.groupby("true_label"):
        class_rows.append({
            "condition": condition,
            "class": cls,
            "n": len(cg),
            "accuracy": cg["correct"].mean()
        })

summary = pd.DataFrame(summary_rows)
paired = pd.DataFrame(paired_rows)
per_class = pd.DataFrame(class_rows)

# Holm multiple-testing correction
pvals = paired["mcnemar_exact_p"].to_numpy()
order = np.argsort(pvals)
adjusted = np.empty(len(pvals))
running = 0

for rank, idx in enumerate(order):
    running = max(running, min(1.0, (len(pvals)-rank)*pvals[idx]))
    adjusted[idx] = running

paired["mcnemar_holm_p"] = adjusted

summary.to_csv(out / "seed42_robustness_summary.csv", index=False)
paired.to_csv(out / "seed42_paired_mcnemar.csv", index=False)
per_class.to_csv(out / "seed42_per_class_metrics.csv", index=False)

plt.figure(figsize=(11, 7))
plot_df = summary.sort_values("stress_accuracy")
plt.barh(plot_df["condition"], plot_df["stress_accuracy"] * 100)
plt.xlabel("Accuracy (%)")
plt.title("Corrected CAFR-Net v2 — Seed 42 Robustness")
plt.tight_layout()
plt.savefig(out / "seed42_robustness_accuracy.png", dpi=200)
plt.close()

metadata = {
    "architecture": "CAFR-Net v2",
    "seed": 42,
    "training_epochs": 20,
    "best_checkpoint_epoch": 2,
    "best_validation_accuracy": 0.9985925404644617,
    "original_test_images": len(original),
    "stress_prediction_rows": len(stress),
    "conditions": len(summary),
    "sample_size_per_condition": sorted(summary["n"].unique().tolist()),
    "matching": "class plus filename stem, extension ignored",
    "statistics": [
        "paired bootstrap 95% confidence intervals",
        "exact McNemar tests",
        "Holm correction"
    ],
    "limitation": "300-image sample per stress condition"
}

(out / "experiment_metadata.json").write_text(json.dumps(metadata, indent=2))

zip_path = out / "CAFRNet_seed42_robustness_results.zip"
with zipfile.ZipFile(zip_path, "w", zipfile.ZIP_DEFLATED) as z:
    for f in out.iterdir():
        if f.is_file() and f != zip_path:
            z.write(f, arcname=f.name)
    z.write(stress_path, arcname="raw/seed42_stress_predictions.csv")
    z.write(base_path, arcname="raw/seed42_original_predictions.csv")

print("PAIRED ANALYSIS COMPLETED")
display(summary.round(4))
print("\nPAIRED STATISTICAL TESTS")
display(paired.round(6))
print("\nFINAL ZIP:", zip_path)
print("ZIP SIZE:", round(zip_path.stat().st_size / 1024**2, 2), "MB")
```

```python
from pathlib import Path
import hashlib

ROOT = Path("/content")
TRAIN = ROOT / "cafr_data" / "train"
VAL = ROOT / "cafr_data" / "val"
OFFICIAL_TEST = ROOT / "imagenette2-320" / "val"
CAFR_TEST = ROOT / "cafr_test"

EXTENSIONS = {".jpg", ".jpeg", ".png", ".webp"}

def image_files(folder):
    return {
        str(p.relative_to(folder)): p
        for p in folder.rglob("*")
        if p.is_file() and p.suffix.lower() in EXTENSIONS
    }

def sha256(path):
    h = hashlib.sha256()
    with open(path, "rb") as f:
        for chunk in iter(lambda: f.read(1024 * 1024), b""):
            h.update(chunk)
    return h.hexdigest()

train_files = image_files(TRAIN)
val_files = image_files(VAL)
official_files = image_files(OFFICIAL_TEST)
cafr_files = image_files(CAFR_TEST)

print("=== IMAGE COUNTS ===")
print("Train:", len(train_files))
print("Validation:", len(val_files))
print("Official test:", len(official_files))
print("CAFR test:", len(cafr_files))

print("\n=== SPLIT OVERLAP CHECK ===")
train_names = set(train_files)
val_names = set(val_files)
test_names = set(official_files)

print("Train/validation overlap:", len(train_names & val_names))
print("Train/test overlap:", len(train_names & test_names))
print("Validation/test overlap:", len(val_names & test_names))

print("\n=== TEST IMAGE IDENTITY ===")
official_names = set(official_files)
cafr_names = set(cafr_files)

print("Same relative paths:", official_names == cafr_names)
print("Missing from CAFR test:", len(official_names - cafr_names))
print("Extra in CAFR test:", len(cafr_names - official_names))

if official_names == cafr_names:
    mismatches = []
    for name in sorted(official_names):
        if sha256(official_files[name]) != sha256(cafr_files[name]):
            mismatches.append(name)

    print("Different image contents:", len(mismatches))
    if mismatches:
        print("Examples:", mismatches[:10])
    else:
        print("PASS: Both folders contain identical original images.")

print("\n=== FINAL VERDICT ===")
if (
    not (train_names & val_names)
    and not (train_names & test_names)
    and not (val_names & test_names)
    and official_names == cafr_names
    and not mismatches
):
    print("PASS: No split overlap; test folders match exactly.")
else:
    print("REVIEW REQUIRED: Check the overlap or identity results.")
```

```python
from pathlib import Path
import subprocess
import torch
import json

REPO = Path("/content/cartoonization-robustness-study")
SOURCE = REPO / "research/CAFRNet_v2/source"

CONVNEXT = Path("/content/cafr_runs/matched_convnext_seed42")
RESNET = Path("/content/cafr_runs/matched_resnet_seed42")

# Verify ConvNeXt output before proceeding
checkpoint = CONVNEXT / "best_model.pt"
history_file = CONVNEXT / "history.json"

print("=== CONVNEXT VERIFICATION ===")

if not checkpoint.exists():
    raise FileNotFoundError(f"ConvNeXt checkpoint missing: {checkpoint}")

ck = torch.load(checkpoint, map_location="cpu", weights_only=False)

print("Architecture:", ck.get("architecture"))
print("Seed:", ck.get("seed"))
print("Best epoch:", ck.get("epoch"))
print("Best validation accuracy:", ck.get("val_accuracy"))

assert ck.get("architecture") == "convnext_tiny"
assert int(ck.get("seed")) == 42

if history_file.exists():
    history = json.loads(history_file.read_text())
    print("Recorded epochs:", len(history))
    assert len(history) == 20, "Expected 20 recorded epochs."

print("\n=== TRAIN RESNET-18 SEED 42 ===")

RESNET.mkdir(parents=True, exist_ok=True)

cmd = [
    "python", "-u", "train_architecture.py",
    "--arch", "resnet18",
    "--data", "/content/cafr_data",
    "--out", str(RESNET),
    "--seed", "42",
    "--epochs", "20",
    "--batch-size", "16",
    "--image-size", "224",
    "--workers", "2",
    "--encoder-lr", "2e-5",
    "--head-lr", "2e-4",
    "--weight-decay", "0.05",
    "--freeze-epochs", "2",
    "--lambda-consistency", "0.25"
]

subprocess.run(cmd, cwd=SOURCE, check=True)

print("\nRESNET-18 SEED 42 TRAINING FINISHED.")
```

```python
import os
import json
import shutil
import zipfile
import subprocess
from pathlib import Path
from getpass import getpass

# --------------------------------------------------
# SETTINGS
# --------------------------------------------------
REPO = "Karthik6-cpu/cartoonization-robustness-study"
BRANCH = "research/cafr-net-phase-iv"
TAG = "cafrnet-v2-drive-backup-2026-11-25"

DRIVE = Path("/content/drive/MyDrive/CAFRNet_v2_Experiments")
WORK = Path("/content/cafr_github_backup")
ZIP_PATH = Path("/content/drive/MyDrive/CAFRNet_v2_Experiments_complete_backup_2026-11-25.zip")

assert DRIVE.exists(), "Google Drive experiment folder not found."

# --------------------------------------------------
# 1. CREATE COMPLETE ZIP BACKUP
# --------------------------------------------------
print("Creating complete ZIP backup...")

with zipfile.ZipFile(ZIP_PATH, "w", zipfile.ZIP_DEFLATED) as z:
    for p in DRIVE.rglob("*"):
        if p.is_file():
            z.write(p, Path("CAFRNet_v2_Experiments") / p.relative_to(DRIVE))

print("ZIP created:", ZIP_PATH)
print("ZIP size:", round(ZIP_PATH.stat().st_size / (1024**2), 2), "MB")

# --------------------------------------------------
# 2. INSTALL GITHUB CLI IF NEEDED
# --------------------------------------------------
if shutil.which("gh") is None:
    subprocess.run(
        ["bash", "-lc",
         "type -p curl >/dev/null || (apt-get update && apt-get install -y curl); "
         "curl -fsSL https://cli.github.com/packages/githubcli-archive-keyring.gpg "
         "| dd of=/usr/share/keyrings/githubcli-archive-keyring.gpg && "
         "chmod go+r /usr/share/keyrings/githubcli-archive-keyring.gpg && "
         "echo 'deb [arch=$(dpkg --print-architecture) "
         "signed-by=/usr/share/keyrings/githubcli-archive-keyring.gpg] "
         "https://cli.github.com/packages stable main' "
         "> /etc/apt/sources.list.d/github-cli.list && "
         "apt-get update && apt-get install -y gh"],
        check=True
    )

# --------------------------------------------------
# 3. AUTHENTICATE
# --------------------------------------------------
token = getpass("Paste your GitHub Personal Access Token: ")

subprocess.run(
    ["gh", "auth", "login", "--hostname", "github.com", "--with-token"],
    input=token + "\n",
    text=True,
    check=True
)
del token

subprocess.run(["gh", "auth", "setup-git"], check=True)

# --------------------------------------------------
# 4. CLONE THE EXISTING RESEARCH BRANCH
# --------------------------------------------------
if WORK.exists():
    shutil.rmtree(WORK)

subprocess.run(
    ["git", "clone", "--branch", BRANCH,
     f"https://github.com/{REPO}.git", str(WORK)],
    check=True
)

# --------------------------------------------------
# 5. COPY DRIVE FILES EXCEPT LARGE CHECKPOINTS
# --------------------------------------------------
DEST = WORK / "research" / "CAFRNet_v2" / "drive_backup_2026-11-25"
DEST.mkdir(parents=True, exist_ok=True)

for p in DRIVE.rglob("*"):
    if p.is_file() and p.suffix.lower() not in [".pt", ".pth"]:
        target = DEST / p.relative_to(DRIVE)
        target.parent.mkdir(parents=True, exist_ok=True)
        shutil.copy2(p, target)

readme = DEST / "README.md"
readme.write_text(
    "# CAFR-Net v2 Drive Backup — 2026-11-25\n\n"
    "This directory preserves the small experiment files recovered "
    "from Google Drive.\n\n"
    "The complete archive, including best_model.pt and latest_model.pt, "
    "is attached to the GitHub Release "
    f"`{TAG}`.\n\n"
    "Original Drive folder: `CAFRNet_v2_Experiments`.\n"
    "The original files were not modified.\n"
)

# --------------------------------------------------
# 6. COMMIT AND PUSH SMALL FILES
# --------------------------------------------------
subprocess.run(["git", "-C", str(WORK), "config",
                "user.name", "CAFR-Net Research Backup"], check=True)
subprocess.run(["git", "-C", str(WORK), "config",
                "user.email", "research-backup@users.noreply.github.com"], check=True)

subprocess.run(["git", "-C", str(WORK), "add",
                "research/CAFRNet_v2/drive_backup_2026-11-25"], check=True)

subprocess.run(
    ["git", "-C", str(WORK), "commit", "-m",
     "Back up CAFR-Net v2 Drive experiment metadata and manifests"],
    check=True
)

subprocess.run(["git", "-C", str(WORK), "push", "origin", BRANCH],
               check=True)

print("SUCCESS: Small files pushed to research branch.")

# --------------------------------------------------
# 7. UPLOAD COMPLETE ZIP AS A GITHUB RELEASE
# --------------------------------------------------
print("Uploading complete ZIP as GitHub Release asset...")

subprocess.run(
    ["gh", "release", "create", TAG, str(ZIP_PATH),
     "--repo", REPO,
     "--title", "CAFR-Net v2 Complete Drive Backup — 25 Nov 2026",
     "--notes",
     "Complete archive of the nine experiment files recovered from "
     "Google Drive, including best_model.pt, latest_model.pt, "
     "training metadata, split manifests, configuration and logs."],
    check=True
)

print("\n" + "=" * 65)
print("BACKUP COMPLETED")
print("Research branch:", BRANCH)
print("Release:", f"https://github.com/{REPO}/releases/tag/{TAG}")
print("Drive ZIP:", ZIP_PATH)
print("=" * 65)
```
