```python
# ============================================
# CAFR-Net | DATASET IDENTITY AUDIT
# ============================================

from pathlib import Path
import hashlib
import json

root = Path("/content/imagenette2-320")

def image_files(folder):
    return sorted(
        p for p in folder.rglob("*")
        if p.suffix.lower() in {".jpg", ".jpeg", ".png"}
    )

train_files = image_files(root / "train")
val_files = image_files(root / "val")

def hash_file(path):
    h = hashlib.sha256()
    with open(path, "rb") as f:
        for chunk in iter(lambda: f.read(1024 * 1024), b""):
            h.update(chunk)
    return h.hexdigest()

print("TRAIN:", len(train_files))
print("VALIDATION:", len(val_files))

train_hashes = {hash_file(p) for p in train_files}
val_hashes = {hash_file(p) for p in val_files}

overlap = train_hashes.intersection(val_hashes)

print("Exact duplicate images across splits:", len(overlap))

assert len(overlap) == 0, "Duplicate images detected!"

print("\nFolder-level duplicate audit passed.")
print("NOTE: This does not establish equivalence with the Phase I manifest.")
```

```python
import os
import hashlib
import pandas as pd
from collections import defaultdict

DATASET_ROOT = "/content/imagenette2-320"
MANIFEST_PATH = "/content/recovered_manifests/split_manifest.csv"

manifest = pd.read_csv(MANIFEST_PATH)

def sha256_file(path):
    h = hashlib.sha256()
    with open(path, "rb") as f:
        for chunk in iter(lambda: f.read(1024 * 1024), b""):
            h.update(chunk)
    return h.hexdigest()

print("Indexing current Imagenette images...")

hash_index = defaultdict(list)

for root, dirs, files in os.walk(DATASET_ROOT):
    for filename in files:
        if filename.lower().endswith((".jpg", ".jpeg", ".png")):
            path = os.path.join(root, filename)
            digest = sha256_file(path)
            hash_index[digest].append(path)

print("Current dataset images indexed:", sum(map(len, hash_index.values())))

print("\nComparing original manifest SHA256 values...")

matched_paths = []
match_status = []

for _, row in manifest.iterrows():
    digest = str(row["sha256"]).strip().lower()
    matches = hash_index.get(digest, [])

    matched_paths.append(matches[0] if len(matches) == 1 else None)

    if len(matches) == 1:
        match_status.append("EXACT_MATCH")
    elif len(matches) == 0:
        match_status.append("MISSING")
    else:
        match_status.append("AMBIGUOUS")

manifest["current_path"] = matched_paths
manifest["hash_match_status"] = match_status

print("\nHASH MATCH REPORT")
print(manifest.groupby(["split", "hash_match_status"]).size())

print("\nOVERALL")
print(manifest["hash_match_status"].value_counts())

print("\nCLASS LABEL CHECK")
matched = manifest[manifest["hash_match_status"] == "EXACT_MATCH"].copy()

matched["current_class"] = matched["current_path"].apply(
    lambda p: os.path.basename(os.path.dirname(p))
)

print("Class mismatches:",
      (matched["label"] != matched["current_class"]).sum())

# Save the verified mapping for future reproducibility
OUTPUT = "/content/original_split_hash_verification.csv"
manifest.to_csv(OUTPUT, index=False)

print("\nVerification report saved:", OUTPUT)
```

```python
import os
import hashlib
import pandas as pd
from pathlib import Path

SOURCE = "/content/imagenette2-320"
MANIFEST = "/content/original_split_hash_verification.csv"
DEST = Path("/content/cafr_data")

df = pd.read_csv(MANIFEST)

# Strict identity checks before creating anything
assert len(df) == 13394
assert df["hash_match_status"].eq("EXACT_MATCH").all()
assert df["current_path"].notna().all()
assert df["label"].notna().all()

expected = {"train": 8048, "validation": 1421, "test": 3925}
actual = df["split"].value_counts().to_dict()
assert actual == expected, f"Unexpected split counts: {actual}"

# Create exact split folders using original manifest assignments
for split in ["train", "validation", "test"]:
    folder_name = "val" if split == "validation" else split

    for label in sorted(df["label"].unique()):
        (DEST / folder_name / label).mkdir(
            parents=True, exist_ok=True
        )

# Create links without modifying source images
created = 0

for row in df.itertuples(index=False):
    split = "val" if row.split == "validation" else row.split
    source = Path(row.current_path)

    assert source.is_file(), f"Missing image: {source}"

    target = DEST / split / row.label / source.name

    if not target.exists():
        os.symlink(source, target)

    created += 1

# Verify folder counts
print("RECONSTRUCTED DATASET")

for split, expected_count in [
    ("train", 8048),
    ("val", 1421),
    ("test", 3925)
]:
    folder = DEST / split
    count = sum(1 for p in folder.rglob("*") if p.is_file())

    print(f"{split}: {count}")
    assert count == expected_count

# Verify class mappings
for split in ["train", "val", "test"]:
    classes = sorted(p.name for p in (DEST / split).iterdir() if p.is_dir())
    assert classes == sorted(df["label"].unique())

# Preserve verification artifacts
df.to_csv(DEST / "verified_split_manifest.csv", index=False)

manifest_hash = hashlib.sha256(
    Path(MANIFEST).read_bytes()
).hexdigest()

(DEST / "manifest_sha256.txt").write_text(
    f"SHA256: {manifest_hash}\n"
)

print("\nSUCCESS: Exact original split reconstructed.")
print("Manifest SHA256:", manifest_hash)
print("Dataset location:", DEST)
```

```python
import os, zipfile, shutil
from pathlib import Path
import pandas as pd
import torch

REPO = Path("/content/cartoonization-robustness-study")
MANIFEST = REPO / "research/manifests/matched_all_four_test_manifest.csv"
OUT = Path("/content/cafr_eval")

archives = {
    "opencv": REPO / "generated_datasets/opencv_full.zip",
    "animegan2": REPO / "generated_datasets/animegan2_full.zip",
    "stable_diffusion": REPO / "generated_datasets/stable_diffusion_full.zip",
}

# Load checkpoint classes and matched manifest
ckpt = torch.load(
    "/content/cafr_runs/seed42_pilot/best_cafr_net.pt",
    map_location="cpu",
    weights_only=False
)
classes = ckpt["classes"]
df = pd.read_csv(MANIFEST)

assert len(df) == 3925, f"Unexpected manifest size: {len(df)}"
assert df["source_id"].nunique() == 3925, "Duplicate source IDs"
assert set(df["class_label"].astype(str)) == set(classes), "Class mismatch"

# Build original-image lookup by source ID
original_lookup = {}
for root, _, files in os.walk("/content/cafr_data/test"):
    for filename in files:
        if Path(filename).suffix.lower() in [".jpg", ".jpeg", ".png"]:
            original_lookup[Path(filename).stem] = Path(root) / filename

assert len(original_lookup) == 3925, (
    f"Expected 3925 original images, found {len(original_lookup)}"
)

# Prepare class folders
for condition in ["original", *archives.keys()]:
    (OUT / condition).mkdir(parents=True, exist_ok=True)

# Link original test images
for row in df.itertuples():
    sid = str(row.source_id)
    label = str(row.class_label)
    src = original_lookup.get(sid)
    if src is None:
        raise FileNotFoundError(f"Original image missing: {sid}")

    dst = OUT / "original" / label / f"{sid}{src.suffix.lower()}"
    dst.parent.mkdir(parents=True, exist_ok=True)
    if not dst.exists():
        os.symlink(src, dst)

print("Original test images linked.")

# Extract cartoon archives into class folders using source_id
for condition, archive in archives.items():
    print(f"\nPreparing {condition}...")
    assert archive.is_file(), f"Missing archive: {archive}"

    target_root = OUT / condition
    with zipfile.ZipFile(archive) as z:
        members = [
            m for m in z.namelist()
            if not m.endswith("/") and Path(m).suffix.lower() in [".png", ".jpg", ".jpeg"]
        ]

        member_lookup = {}
        for member in members:
            sid = Path(member).stem
            if sid in member_lookup:
                raise ValueError(f"Duplicate archive identity: {sid}")
            member_lookup[sid] = member

        expected = set(df["source_id"].astype(str))
        actual = set(member_lookup)

        if expected != actual:
            raise ValueError(
                f"{condition} identity mismatch: "
                f"missing={len(expected-actual)}, extra={len(actual-expected)}"
            )

        for row in df.itertuples():
            sid = str(row.source_id)
            label = str(row.class_label)
            member = member_lookup[sid]
            suffix = Path(member).suffix.lower()
            dst = target_root / label / f"{sid}{suffix}"
            dst.parent.mkdir(parents=True, exist_ok=True)

            if not dst.exists():
                with z.open(member) as src, open(dst, "wb") as out:
                    shutil.copyfileobj(src, out)

    count = sum(
        len(files)
        for _, _, files in os.walk(target_root)
    )
    assert count == 3925, f"{condition}: expected 3925, got {count}"
    print(f"{condition}: {count} images verified")

# Final verification
print("\n========== FINAL TEST FOLDER AUDIT ==========")
for condition in ["original", *archives.keys()]:
    count = sum(len(files) for _, _, files in os.walk(OUT / condition))
    print(f"{condition:20s}: {count}")
    assert count == 3925

print("\nSUCCESS: Four matched test conditions prepared.")
print("Test folder:", OUT)
```

```python
import os, json, zipfile
from pathlib import Path

import numpy as np
import pandas as pd
from sklearn.metrics import accuracy_score, f1_score, recall_score, confusion_matrix
from scipy.stats import binomtest

REPO = Path("/content/cartoonization-robustness-study")
RUN = Path("/content/cafr_runs/seed42_pilot")
CSV = RUN / "test_predictions.csv"
OUT = RUN / "final_evaluation"
OUT.mkdir(parents=True, exist_ok=True)

df = pd.read_csv(CSV)
required = {
    "condition", "relative_path", "true_label",
    "predicted_label", "confidence", "correct"
}
assert required.issubset(df.columns), f"Missing columns: {required-set(df.columns)}"

df["source_id"] = df["relative_path"].map(lambda x: Path(str(x)).stem)
df["correct"] = df["correct"].astype(int)

conditions = ["original", "opencv", "animegan2", "stable_diffusion"]

# Validate matched identities and labels
assert len(df) == 15700, f"Expected 15,700 rows, got {len(df)}"
assert set(df["condition"]) == set(conditions)

for c in conditions:
    part = df[df.condition == c]
    assert len(part) == 3925, f"{c}: wrong row count"
    assert part.source_id.nunique() == 3925, f"{c}: duplicate identities"

wide = df.pivot(
    index="source_id",
    columns="condition",
    values=["true_label", "predicted_label", "correct", "confidence"]
)
assert len(wide) == 3925
assert wide["true_label"].nunique(axis=1).max() == 1

# Calibration metric
def ece_score(y_true, y_pred, confidence, bins=15):
    correct = (np.asarray(y_true) == np.asarray(y_pred)).astype(float)
    confidence = np.asarray(confidence)
    edges = np.linspace(0, 1, bins + 1)
    result = 0.0
    for i in range(bins):
        mask = (
            (confidence >= edges[i]) &
            (confidence <= edges[i+1] if i == bins-1 else confidence < edges[i+1])
        )
        if mask.any():
            result += mask.mean() * abs(correct[mask].mean() - confidence[mask].mean())
    return float(result)

# Per-condition metrics
summary = []
recall_rows = []
confusion_rows = []

for c in conditions:
    part = df[df.condition == c].copy()
    y = part.true_label.astype(str)
    pred = part.predicted_label.astype(str)

    summary.append({
        "condition": c,
        "n": len(part),
        "top1_accuracy": accuracy_score(y, pred),
        "macro_f1": f1_score(y, pred, average="macro", zero_division=0),
        "ece_15_bins": ece_score(y, pred, part.confidence),
        "mean_confidence": part.confidence.mean(),
        "prediction_error_rate": 1 - part.correct.mean()
    })

    recalls = recall_score(y, pred, labels=sorted(y.unique()),
                           average=None, zero_division=0)
    for label, value in zip(sorted(y.unique()), recalls):
        recall_rows.append({
            "condition": c, "class": label, "recall": value
        })

    cm = confusion_matrix(y, pred, labels=sorted(y.unique()))
    for i, label in enumerate(sorted(y.unique())):
        for j, predicted in enumerate(sorted(y.unique())):
            confusion_rows.append({
                "condition": c,
                "true_class": label,
                "predicted_class": predicted,
                "count": int(cm[i, j])
            })

summary_df = pd.DataFrame(summary)
recall_df = pd.DataFrame(recall_rows)
confusion_df = pd.DataFrame(confusion_rows)

# Paired bootstrap: cartoon condition accuracy minus original accuracy
rng = np.random.default_rng(42)
original_correct = wide["correct"]["original"].to_numpy(dtype=int)
paired_rows = []

for c in conditions[1:]:
    condition_correct = wide["correct"][c].to_numpy(dtype=int)
    delta = condition_correct - original_correct
    n = len(delta)

    boot = np.empty(10000)
    for b in range(len(boot)):
        idx = rng.integers(0, n, n)
        boot[b] = delta[idx].mean()

    # Exact paired McNemar test
    original_only = int(((original_correct == 1) & (condition_correct == 0)).sum())
    condition_only = int(((original_correct == 0) & (condition_correct == 1)).sum())
    discordant = original_only + condition_only
    p = binomtest(
        min(original_only, condition_only),
        n=discordant,
        p=0.5,
        alternative="two-sided"
    ).pvalue if discordant else 1.0

    paired_rows.append({
        "comparison": f"{c} vs original",
        "original_accuracy": original_correct.mean(),
        "condition_accuracy": condition_correct.mean(),
        "accuracy_difference": delta.mean(),
        "bootstrap_95ci_low": np.quantile(boot, 0.025),
        "bootstrap_95ci_high": np.quantile(boot, 0.975),
        "original_correct_condition_wrong": original_only,
        "original_wrong_condition_correct": condition_only,
        "mcnemar_p_raw": p,
        "prediction_flip_rate": (
            wide["predicted_label"][c].to_numpy() !=
            wide["predicted_label"]["original"].to_numpy()
        ).mean()
    })

paired_df = pd.DataFrame(paired_rows)

# Holm multiple-comparison correction
pvals = paired_df["mcnemar_p_raw"].to_numpy()
order = np.argsort(pvals)
adjusted = np.ones(len(pvals))
running_max = 0.0
m = len(pvals)

for rank, idx in enumerate(order):
    adjusted[idx] = min(1.0, (m-rank) * pvals[idx])
    running_max = max(running_max, adjusted[idx])
    adjusted[idx] = running_max

paired_df["mcnemar_p_holm"] = adjusted

# Save outputs
summary_df.to_csv(OUT / "condition_metrics.csv", index=False)
recall_df.to_csv(OUT / "per_class_recall.csv", index=False)
confusion_df.to_csv(OUT / "confusion_matrices.csv", index=False)
paired_df.to_csv(OUT / "paired_statistics.csv", index=False)

report = {
    "model": "CAFR-Net",
    "checkpoint": str(RUN / "best_cafr_net.pt"),
    "test_images_per_condition": 3925,
    "total_prediction_rows": len(df),
    "conditions": conditions,
    "seed": 42,
    "interpretation": (
        "Single-seed pilot evaluation. These results do not establish "
        "multi-seed robustness or superiority over baselines."
    )
}
(OUT / "evaluation_metadata.json").write_text(json.dumps(report, indent=2))

print("\n========== CONDITION METRICS ==========")
print(summary_df.to_string(index=False))

print("\n========== PAIRED STATISTICS ==========")
print(paired_df.to_string(index=False))

print("\n========== PER-CLASS RECALL ==========")
print(recall_df.pivot(index="class", columns="condition",
                      values="recall").round(4).to_string())

# Build reproducibility package
package = RUN / "CAFRNet_PhaseIV_Pilot_Results.zip"
include = [
    RUN / "best_cafr_net.pt",
    RUN / "latest_cafr_net.pt",
    RUN / "test_predictions.csv",
    RUN / "history.json",
    RUN / "run_metadata.json",
    REPO / "models/cafr_net.py",
    REPO / "training/train_cafr_net.py",
    REPO / "evaluation/evaluate_cafr_net.py",
    REPO / "research_phase_iv/PROTOCOL.md",
    REPO / "research_phase_iv/config.yaml",
]

with zipfile.ZipFile(package, "w", zipfile.ZIP_DEFLATED) as z:
    for p in include:
        if p.exists():
            z.write(p, arcname=str(p.relative_to(REPO.parent)))
    for p in OUT.glob("*"):
        if p.is_file():
            z.write(p, arcname=f"final_evaluation/{p.name}")

print("\nRESULT FILES:", OUT)
print("RESEARCH PACKAGE:", package)
print("PACKAGE SIZE MB:", round(package.stat().st_size / 1024**2, 2))
```

```python
%cd /content/cartoonization-robustness-study

import os, json, random, io
import numpy as np
import torch
import torch.nn as nn
from PIL import Image, ImageEnhance, ImageFilter
from torch.utils.data import Dataset, DataLoader
from torchvision import transforms
from torchvision.datasets import ImageFolder
from torchvision.models import convnext_tiny, ConvNeXt_Tiny_Weights

# Reproducibility
SEED = 42
random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
torch.cuda.manual_seed_all(SEED)

DEVICE = "cuda" if torch.cuda.is_available() else "cpu"
DATA = "/content/cafr_data"
OUT = "/content/cafr_runs/convnext_baseline_seed42"
os.makedirs(OUT, exist_ok=True)

MEAN = (0.485, 0.456, 0.406)
STD = (0.229, 0.224, 0.225)

def shifted_view(image):
    kind = random.choice(("jpeg", "blur", "brightness", "noise", "identity"))
    if kind == "blur":
        return image.filter(ImageFilter.GaussianBlur(random.uniform(.3, 1.2)))
    if kind == "brightness":
        return ImageEnhance.Brightness(image).enhance(random.uniform(.75, 1.25))
    if kind == "jpeg":
        b = io.BytesIO()
        image.save(b, format="JPEG", quality=random.choice((35, 50, 65, 80)))
        b.seek(0)
        return Image.open(b).convert("RGB")
    if kind == "noise":
        arr = np.asarray(image).astype(np.float32)
        arr += np.random.normal(0, random.uniform(3, 12), arr.shape)
        return Image.fromarray(np.uint8(np.clip(arr, 0, 255)))
    return image.copy()

class PairedDataset(Dataset):
    def __init__(self, root):
        self.base = ImageFolder(root)
        self.geometry = transforms.Compose([
            transforms.Resize((224, 224)),
            transforms.RandomHorizontalFlip(.5)
        ])
        self.tensor = transforms.Compose([
            transforms.ToTensor(),
            transforms.Normalize(MEAN, STD)
        ])

    def __len__(self):
        return len(self.base)

    def __getitem__(self, idx):
        path, label = self.base.samples[idx]
        image = self.base.loader(path).convert("RGB")
        clean = self.geometry(image)
        shifted = shifted_view(clean)
        return self.tensor(clean), self.tensor(shifted), label

class ConvNeXtBaseline(nn.Module):
    def __init__(self, num_classes):
        super().__init__()
        self.net = convnext_tiny(weights=ConvNeXt_Tiny_Weights.DEFAULT)
        self.net.classifier[2] = nn.Linear(
            self.net.classifier[2].in_features, num_classes
        )

    def forward(self, x):
        return self.net(x)

train_ds = PairedDataset(DATA + "/train")
val_base = ImageFolder(DATA + "/val")
assert train_ds.base.classes == val_base.classes

val_tf = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize(MEAN, STD)
])
val_ds = ImageFolder(DATA + "/val", transform=val_tf)

train_loader = DataLoader(
    train_ds, batch_size=16, shuffle=True,
    num_workers=2, pin_memory=True
)
val_loader = DataLoader(
    val_ds, batch_size=32, shuffle=False,
    num_workers=2, pin_memory=True
)

model = ConvNeXtBaseline(len(train_ds.base.classes)).to(DEVICE)
optimizer = torch.optim.AdamW(model.parameters(), lr=2e-5, weight_decay=.05)
scheduler = torch.optim.lr_scheduler.CosineAnnealingLR(optimizer, T_max=5)
criterion = nn.CrossEntropyLoss()
scaler = torch.amp.GradScaler("cuda", enabled=DEVICE == "cuda")

history = []
best_acc = 0.0

for epoch in range(5):
    model.train()
    total_loss = 0.0
    total = 0

    for clean, shifted, labels in train_loader:
        clean = clean.to(DEVICE)
        shifted = shifted.to(DEVICE)
        labels = labels.to(DEVICE)

        optimizer.zero_grad(set_to_none=True)

        with torch.autocast(device_type="cuda", enabled=DEVICE == "cuda"):
            logits_a = model(clean)
            logits_b = model(shifted)
            loss = criterion(logits_a, labels) + criterion(logits_b, labels)

        scaler.scale(loss).backward()
        scaler.step(optimizer)
        scaler.update()

        total_loss += loss.item() * labels.size(0)
        total += labels.size(0)

    model.eval()
    correct = 0
    count = 0
    val_loss = 0.0

    with torch.no_grad():
        for images, labels in val_loader:
            images, labels = images.to(DEVICE), labels.to(DEVICE)
            with torch.autocast(device_type="cuda", enabled=DEVICE == "cuda"):
                logits = model(images)
                loss = criterion(logits, labels)

            val_loss += loss.item() * labels.size(0)
            correct += (logits.argmax(1) == labels).sum().item()
            count += labels.size(0)

    scheduler.step()
    accuracy = correct / count
    row = {
        "epoch": epoch + 1,
        "train_loss": total_loss / total,
        "val_loss": val_loss / count,
        "val_accuracy": accuracy
    }
    history.append(row)
    print(json.dumps(row), flush=True)

    if accuracy > best_acc:
        best_acc = accuracy
        torch.save({
            "model": model.state_dict(),
            "classes": train_ds.base.classes,
            "epoch": epoch + 1,
            "val_accuracy": accuracy,
            "seed": SEED
        }, OUT + "/best_convnext_baseline.pt")

    torch.save({
        "model": model.state_dict(),
        "optimizer": optimizer.state_dict(),
        "scheduler": scheduler.state_dict(),
        "epoch": epoch + 1,
        "history": history
    }, OUT + "/latest_convnext_baseline.pt")

with open(OUT + "/history.json", "w") as f:
    json.dump(history, f, indent=2)

print("\nConvNeXt baseline pilot completed.")
print("Best validation accuracy:", best_acc)
print("Checkpoint:", OUT + "/best_convnext_baseline.pt")
```

```python
import torch
import pandas as pd
from pathlib import Path
from torch import nn
from torch.utils.data import DataLoader
from torchvision import transforms
from torchvision.datasets import ImageFolder
from torchvision.models import convnext_tiny

DEVICE = "cuda" if torch.cuda.is_available() else "cpu"

CKPT_PATH = "/content/cafr_runs/convnext_baseline_seed42/best_convnext_baseline.pt"
OUT_PATH = "/content/cafr_runs/convnext_baseline_seed42/test_predictions.csv"

ckpt = torch.load(CKPT_PATH, map_location="cpu", weights_only=False)
classes = ckpt["classes"]

# Recreate the exact training model structure
class ConvNeXtBaseline(nn.Module):
    def __init__(self, num_classes):
        super().__init__()
        self.net = convnext_tiny(weights=None)
        self.net.classifier[2] = nn.Linear(
            self.net.classifier[2].in_features, num_classes
        )

    def forward(self, x):
        return self.net(x)

model = ConvNeXtBaseline(len(classes))
model.load_state_dict(ckpt["model"], strict=True)
model.to(DEVICE).eval()

print("Checkpoint loaded successfully.")
print("Best validation accuracy:", ckpt["val_accuracy"])

tf = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize(
        (0.485, 0.456, 0.406),
        (0.229, 0.224, 0.225)
    )
])

conditions = {
    "original": "/content/cafr_eval/original",
    "opencv": "/content/cafr_eval/opencv",
    "animegan2": "/content/cafr_eval/animegan2",
    "stable_diffusion": "/content/cafr_eval/stable_diffusion"
}

rows = []

with torch.no_grad():
    for condition, path in conditions.items():
        ds = ImageFolder(path, transform=tf)
        assert ds.classes == classes
        assert len(ds) == 3925

        loader = DataLoader(
            ds, batch_size=32, shuffle=False, num_workers=2,
            pin_memory=True
        )

        offset = 0
        for images, labels in loader:
            images = images.to(DEVICE)

            with torch.autocast(
                device_type="cuda",
                enabled=DEVICE == "cuda"
            ):
                logits = model(images)

            probs = logits.float().softmax(1).cpu()
            confidence, predictions = probs.max(1)

            for j in range(len(labels)):
                relative_path = str(
                    Path(ds.samples[offset+j][0]).relative_to(path)
                )
                rows.append({
                    "condition": condition,
                    "relative_path": relative_path,
                    "true_label": classes[int(labels[j])],
                    "predicted_label": classes[int(predictions[j])],
                    "confidence": float(confidence[j]),
                    "correct": int(predictions[j] == labels[j])
                })

            offset += len(labels)

        print(f"{condition}: {offset} predictions completed")

result = pd.DataFrame(rows)
assert len(result) == 15700

result.to_csv(OUT_PATH, index=False)

print("\nSUCCESS: ConvNeXt-Tiny evaluation completed.")
print("Predictions:", len(result))
print("Saved:", OUT_PATH)
```

```python
import numpy as np
import pandas as pd
from pathlib import Path
from scipy.stats import binomtest
from sklearn.metrics import accuracy_score, f1_score

RUN = Path("/content/cafr_runs")
OUT = RUN / "CAFRNet_vs_ConvNeXt_comparison"
OUT.mkdir(parents=True, exist_ok=True)

cafr = pd.read_csv(
    RUN / "seed42_pilot/test_predictions.csv"
)
conv = pd.read_csv(
    RUN / "convnext_baseline_seed42/test_predictions.csv"
)

def prepare(df):
    df = df.copy()
    df["source_id"] = df["relative_path"].map(
        lambda x: Path(str(x)).stem
    )
    return df

cafr = prepare(cafr)
conv = prepare(conv)

conditions = ["original", "opencv", "animegan2", "stable_diffusion"]

# Strict identity and label matching
key = ["source_id", "condition"]

assert len(cafr) == 15700 and len(conv) == 15700
assert not cafr.duplicated(key).any()
assert not conv.duplicated(key).any()

merged = cafr.merge(
    conv,
    on=key,
    suffixes=("_cafr", "_conv"),
    validate="one_to_one"
)

assert len(merged) == 15700
assert (merged.true_label_cafr == merged.true_label_conv).all()

# Expected calibration error
def calculate_ece(y, pred, confidence, bins=15):
    correct = (np.asarray(y) == np.asarray(pred)).astype(float)
    confidence = np.asarray(confidence)
    edges = np.linspace(0, 1, bins + 1)
    ece = 0.0

    for i in range(bins):
        mask = (
            (confidence >= edges[i]) &
            (confidence <= edges[i+1] if i == bins-1
             else confidence < edges[i+1])
        )
        if mask.any():
            ece += mask.mean() * abs(
                correct[mask].mean() - confidence[mask].mean()
            )
    return ece

summary = []
paired = []
rng = np.random.default_rng(42)

for condition in conditions:
    d = merged[merged.condition == condition].copy()

    y = d.true_label_cafr.astype(str).to_numpy()
    p_cafr = d.predicted_label_cafr.astype(str).to_numpy()
    p_conv = d.predicted_label_conv.astype(str).to_numpy()

    correct_cafr = (p_cafr == y).astype(int)
    correct_conv = (p_conv == y).astype(int)

    acc_cafr = correct_cafr.mean()
    acc_conv = correct_conv.mean()

    summary.append({
        "condition": condition,
        "n": len(d),
        "CAFR_accuracy": acc_cafr,
        "ConvNeXt_accuracy": acc_conv,
        "accuracy_difference_pp": 100 * (acc_cafr - acc_conv),
        "CAFR_macro_F1": f1_score(y, p_cafr, average="macro"),
        "ConvNeXt_macro_F1": f1_score(y, p_conv, average="macro"),
        "CAFR_ECE": calculate_ece(
            y, p_cafr, d.confidence_cafr
        ),
        "ConvNeXt_ECE": calculate_ece(
            y, p_conv, d.confidence_conv
        ),
        "CAFR_prediction_flip_rate": (
            p_cafr != p_conv
        ).mean()
    })

    # Paired bootstrap over identical source IDs
    delta = correct_cafr - correct_conv
    n = len(delta)

    boot = np.empty(10000)
    for b in range(len(boot)):
        idx = rng.integers(0, n, n)
        boot[b] = delta[idx].mean()

    # Exact paired McNemar test
    cafr_only = int(
        ((correct_cafr == 1) & (correct_conv == 0)).sum()
    )
    conv_only = int(
        ((correct_cafr == 0) & (correct_conv == 1)).sum()
    )
    discordant = cafr_only + conv_only

    pvalue = (
        binomtest(
            min(cafr_only, conv_only),
            n=discordant,
            p=0.5
        ).pvalue if discordant else 1.0
    )

    paired.append({
        "condition": condition,
        "CAFR_only_correct": cafr_only,
        "ConvNeXt_only_correct": conv_only,
        "discordant_pairs": discordant,
        "accuracy_difference_pp": 100 * delta.mean(),
        "bootstrap_95CI_low_pp": 100 * np.quantile(boot, .025),
        "bootstrap_95CI_high_pp": 100 * np.quantile(boot, .975),
        "mcnemar_p_raw": pvalue
    })

summary = pd.DataFrame(summary)
paired = pd.DataFrame(paired)

# Holm correction
pvals = paired.mcnemar_p_raw.to_numpy()
order = np.argsort(pvals)
adjusted = np.ones(len(pvals))
running_max = 0.0
m = len(pvals)

for rank, idx in enumerate(order):
    value = min(1.0, (m-rank) * pvals[idx])
    running_max = max(running_max, value)
    adjusted[idx] = running_max

paired["mcnemar_p_holm"] = adjusted

summary.to_csv(OUT / "model_comparison_metrics.csv", index=False)
paired.to_csv(OUT / "paired_model_statistics.csv", index=False)

print("\n========== MODEL PERFORMANCE ==========")
print(summary.round(5).to_string(index=False))

print("\n========== PAIRED STATISTICAL TESTS ==========")
print(paired.round(6).to_string(index=False))

print("\nSaved comparison files to:", OUT)
```

```python
%%writefile /content/cartoonization-robustness-study/models/cafr_net_v2.py

import torch
import torch.nn as nn
import torch.nn.functional as F

from torchvision.models import (
    convnext_tiny,
    ConvNeXt_Tiny_Weights,
)


class ConvBNAct(nn.Sequential):
    def __init__(self, in_channels, out_channels, kernel_size=1):
        padding = kernel_size // 2
        super().__init__(
            nn.Conv2d(
                in_channels, out_channels,
                kernel_size, padding=padding, bias=False
            ),
            nn.BatchNorm2d(out_channels),
            nn.GELU()
        )


class ChannelSpatialAttention(nn.Module):
    def __init__(self, channels, reduction=16):
        super().__init__()

        hidden = max(channels // reduction, 8)

        self.channel_mlp = nn.Sequential(
            nn.Conv2d(channels, hidden, 1),
            nn.GELU(),
            nn.Conv2d(hidden, channels, 1)
        )

        self.spatial = nn.Conv2d(
            2, 1, kernel_size=7, padding=3
        )

    def forward(self, x):
        avg = F.adaptive_avg_pool2d(x, 1)
        mx = F.adaptive_max_pool2d(x, 1)

        channel_attention = torch.sigmoid(
            self.channel_mlp(avg) +
            self.channel_mlp(mx)
        )

        x = x * channel_attention

        avg_map = x.mean(dim=1, keepdim=True)
        max_map = x.max(dim=1, keepdim=True)[0]

        spatial_attention = torch.sigmoid(
            self.spatial(torch.cat([avg_map, max_map], dim=1))
        )

        return x * spatial_attention


class SemanticScaleFusion(nn.Module):
    def __init__(self, channels):
        super().__init__()

        self.scale_scorer = nn.Sequential(
            nn.Linear(channels * 3, channels),
            nn.GELU(),
            nn.Dropout(0.1),
            nn.Linear(channels, 3)
        )

    def forward(self, features):
        pooled = [
            F.adaptive_avg_pool2d(f, 1).flatten(1)
            for f in features
        ]

        descriptor = torch.cat(pooled, dim=1)
        weights = torch.softmax(
            self.scale_scorer(descriptor), dim=1
        )

        target_size = features[0].shape[-2:]

        aligned = [
            f if f.shape[-2:] == target_size
            else F.interpolate(
                f, size=target_size,
                mode="bilinear", align_corners=False
            )
            for f in features
        ]

        fused = sum(
            aligned[i] * weights[:, i].view(-1, 1, 1, 1)
            for i in range(3)
        )

        return fused, weights


class EdgeRefinement(nn.Module):
    def __init__(self, channels):
        super().__init__()

        sobel_x = torch.tensor(
            [[-1, 0, 1], [-2, 0, 2], [-1, 0, 1]],
            dtype=torch.float32
        ).view(1, 1, 3, 3)

        sobel_y = torch.tensor(
            [[-1, -2, -1], [0, 0, 0], [1, 2, 1]],
            dtype=torch.float32
        ).view(1, 1, 3, 3)

        self.register_buffer("sobel_x", sobel_x)
        self.register_buffer("sobel_y", sobel_y)

        self.edge_encoder = nn.Sequential(
            ConvBNAct(1, channels, 3),
            ConvBNAct(channels, channels, 3)
        )

        # Bounded residual contribution.
        self.edge_strength = nn.Parameter(torch.tensor(-2.0))

    def forward(self, image, features):
        gray = (
            0.299 * image[:, 0:1] +
            0.587 * image[:, 1:2] +
            0.114 * image[:, 2:3]
        )

        gx = F.conv2d(gray, self.sobel_x, padding=1)
        gy = F.conv2d(gray, self.sobel_y, padding=1)

        edge = torch.sqrt(gx.square() + gy.square() + 1e-6)

        edge = F.interpolate(
            edge,
            size=features.shape[-2:],
            mode="bilinear",
            align_corners=False
        )

        edge_features = self.edge_encoder(edge)

        strength = 0.25 * torch.sigmoid(self.edge_strength)

        return features + strength * edge_features


class CAFRNetV2(nn.Module):
    def __init__(
        self,
        num_classes=10,
        pretrained=True,
        dropout=0.25,
        projection_dim=256
    ):
        super().__init__()

        weights = (
            ConvNeXt_Tiny_Weights.DEFAULT
            if pretrained else None
        )

        backbone = convnext_tiny(weights=weights)
        self.backbone_features = backbone.features

        # ConvNeXt-Tiny feature stages:
        # 3 -> 192 channels
        # 5 -> 384 channels
        # 7 -> 768 channels
        self.stage_indices = (3, 5, 7)

        self.projections = nn.ModuleList([
            ConvBNAct(192, projection_dim),
            ConvBNAct(384, projection_dim),
            ConvBNAct(768, projection_dim)
        ])

        self.fusion = SemanticScaleFusion(projection_dim)

        self.edge_refinement = EdgeRefinement(projection_dim)

        self.attention = ChannelSpatialAttention(projection_dim)

        self.refinement = nn.Sequential(
            ConvBNAct(projection_dim, projection_dim, 3),
            nn.Dropout2d(0.1),
            ConvBNAct(projection_dim, projection_dim, 3)
        )

        self.classifier = nn.Sequential(
            nn.LayerNorm(projection_dim),
            nn.Linear(projection_dim, 512),
            nn.GELU(),
            nn.Dropout(dropout),
            nn.Linear(512, num_classes)
        )

        self.embedding_dim = projection_dim

    def forward(self, x, return_aux=False):
        input_image = x

        stage_features = []
        for index, layer in enumerate(self.backbone_features):
            x = layer(x)

            if index in self.stage_indices:
                stage_features.append(x)

        projected = [
            projection(feature)
            for projection, feature
            in zip(self.projections, stage_features)
        ]

        fused, scale_weights = self.fusion(projected)

        fused = self.edge_refinement(input_image, fused)

        attended = self.attention(fused)

        refined = attended + self.refinement(attended)

        embedding = F.adaptive_avg_pool2d(
            refined, 1
        ).flatten(1)

        logits = self.classifier(embedding)

        if return_aux:
            return {
                "logits": logits,
                "embedding": embedding,
                "scale_weights": scale_weights
            }

        return logits


def paired_consistency_loss(
    logits_a,
    logits_b,
    temperature=2.0
):
    log_p = F.log_softmax(
        logits_a / temperature, dim=1
    )
    log_q = F.log_softmax(
        logits_b / temperature, dim=1
    )

    p = log_p.exp()
    q = log_q.exp()

    loss_ab = F.kl_div(
        log_p, q, reduction="batchmean"
    )

    loss_ba = F.kl_div(
        log_q, p, reduction="batchmean"
    )

    return (
        0.5 * (loss_ab + loss_ba)
        * temperature ** 2
    )


# Compatibility name for existing imports.
CAFRNet = CAFRNetV2
```

```python
import pandas as pd
import numpy as np
from sklearn.metrics import accuracy_score, f1_score

p = "/content/cafr_runs/cafr_v2_opencv_seed42/test_predictions.csv"
df = pd.read_csv(p)

results = []

for condition, d in df.groupby("condition"):
    y = d["true_label"]
    pred = d["predicted_label"]
    confidence = d["confidence"].to_numpy()
    correct = (y.to_numpy() == pred.to_numpy()).astype(int)

    # Expected Calibration Error (15 bins)
    ece = 0.0
    bins = np.linspace(0, 1, 16)

    for i in range(15):
        mask = (
            (confidence >= bins[i]) &
            (confidence < bins[i + 1])
        )
        if mask.any():
            ece += mask.mean() * abs(
                correct[mask].mean() - confidence[mask].mean()
            )

    results.append({
        "condition": condition,
        "samples": len(d),
        "accuracy": accuracy_score(y, pred),
        "macro_f1": f1_score(y, pred, average="macro"),
        "ece": ece
    })

metrics = pd.DataFrame(results)
print(metrics.round(5).to_string(index=False))

out = "/content/cafr_runs/cafr_v2_opencv_seed42"
metrics.to_csv(f"{out}/test_metrics.csv", index=False)

print("\nMetrics saved.")
```

```python
import pandas as pd
import numpy as np
from pathlib import Path
from scipy.stats import binomtest

base = Path("/content/cafr_runs")

files = {
    "CAFR_v1": base / "seed42_pilot/test_predictions.csv",
    "ConvNeXt": base / "convnext_baseline_seed42/test_predictions.csv",
    "CAFR_v2": base / "cafr_v2_opencv_seed42/test_predictions.csv",
}

data = {}

for name, path in files.items():
    df = pd.read_csv(path)
    df = df[df.condition == "opencv"].copy()
    df["source_id"] = df["relative_path"].map(
        lambda x: Path(str(x)).stem
    )
    data[name] = df

v2 = data["CAFR_v2"]

for baseline_name in ["CAFR_v1", "ConvNeXt"]:
    other = data[baseline_name]

    m = v2.merge(
        other,
        on="source_id",
        suffixes=("_v2", "_base"),
        validate="one_to_one"
    )

    assert len(m) == 3925
    assert (m.true_label_v2 == m.true_label_base).all()

    a = (
        m.predicted_label_v2 == m.true_label_v2
    ).to_numpy()

    b = (
        m.predicted_label_base == m.true_label_base
    ).to_numpy()

    diff = a.astype(int) - b.astype(int)
    delta = diff.mean() * 100

    # Paired bootstrap confidence interval
    rng = np.random.default_rng(42)
    boot = []

    for _ in range(10000):
        idx = rng.integers(0, len(diff), len(diff))
        boot.append(diff[idx].mean() * 100)

    low, high = np.percentile(boot, [2.5, 97.5])

    v2_only = int((a & ~b).sum())
    base_only = int((~a & b).sum())

    p = binomtest(
        v2_only,
        v2_only + base_only,
        0.5,
        alternative="two-sided"
    ).pvalue

    print(f"\nCAFR-v2 vs {baseline_name}")
    print("V2 accuracy:", round(a.mean() * 100, 3))
    print("Baseline accuracy:", round(b.mean() * 100, 3))
    print("Accuracy difference (pp):", round(delta, 3))
    print("Paired bootstrap 95% CI:", round(low, 3), "to", round(high, 3))
    print("V2-only correct:", v2_only)
    print("Baseline-only correct:", base_only)
    print("McNemar exact p-value:", p)
```

```python
import pandas as pd
import numpy as np
from pathlib import Path
from sklearn.metrics import accuracy_score, f1_score

base = Path("/content/cafr_runs")

files = {
    "Seed_42": base / "cafr_v2_opencv_seed42/test_predictions.csv",
    "Seed_123": base / "cafr_v2_opencv_seed123/test_predictions.csv"
}

def calculate_ece(y, pred, confidence, bins=15):
    correct = (np.array(y) == np.array(pred)).astype(int)
    confidence = np.array(confidence)
    ece = 0.0

    for i in range(bins):
        lower, upper = i / bins, (i + 1) / bins
        mask = (confidence > lower) & (confidence <= upper)

        if mask.sum():
            ece += mask.mean() * abs(
                correct[mask].mean() - confidence[mask].mean()
            )

    return ece

rows = []

for seed, path in files.items():
    df = pd.read_csv(path)

    print(seed, "rows:", len(df))
    print("Columns:", df.columns.tolist())

    for condition, d in df.groupby("condition"):
        row = {
            "Seed": seed,
            "Condition": condition,
            "Accuracy": accuracy_score(
                d.true_label, d.predicted_label
            ) * 100,
            "Macro_F1": f1_score(
                d.true_label, d.predicted_label,
                average="macro"
            ) * 100
        }

        if "confidence" in d.columns:
            row["ECE"] = calculate_ece(
                d.true_label, d.predicted_label, d.confidence
            )

        rows.append(row)

results = pd.DataFrame(rows)

print("\n===== SEED COMPARISON =====")
for metric in ["Accuracy", "Macro_F1", "ECE"]:
    if metric in results.columns:
        print(f"\n{metric}")
        print(results.pivot(
            index="Condition", columns="Seed", values=metric
        ).round(4).to_string())

results.to_csv(
    base / "CAFRNet_Independent_Seed_Comparison.csv",
    index=False
)

print("\nSaved: CAFRNet_Independent_Seed_Comparison.csv")
```

```python
import pandas as pd
import numpy as np
from pathlib import Path

base = Path("/content/cafr_runs")

s42 = pd.read_csv(
    base / "cafr_v2_opencv_seed42/test_predictions.csv"
)
s123 = pd.read_csv(
    base / "cafr_v2_opencv_seed123/test_predictions.csv"
)

# Compare predictions for matching test images
a = s42[s42.condition == "opencv"].copy()
b = s123[s123.condition == "opencv"].copy()

merged = a.merge(
    b,
    on=["condition", "relative_path", "true_label"],
    suffixes=("_42", "_123"),
    validate="one_to_one"
)

print("Matched OpenCV test images:", len(merged))

# Per-class comparison
rows = []

for label, d in merged.groupby("true_label"):
    rows.append({
        "Class": label,
        "Seed42_Accuracy": d.correct_42.mean() * 100,
        "Seed123_Accuracy": d.correct_123.mean() * 100,
        "Difference_pp": (
            d.correct_123.mean() - d.correct_42.mean()
        ) * 100
    })

class_results = pd.DataFrame(rows)

print("\n===== OPENCV CLASS-WISE COMPARISON =====")
print(
    class_results.sort_values("Difference_pp")
    .round(3).to_string(index=False)
)

print("\n===== PREDICTION AGREEMENT =====")
print("Both correct:", ((merged.correct_42 == 1) &
                        (merged.correct_123 == 1)).sum())
print("Both wrong:", ((merged.correct_42 == 0) &
                      (merged.correct_123 == 0)).sum())
print("Only seed 42 correct:",
      ((merged.correct_42 == 1) &
       (merged.correct_123 == 0)).sum())
print("Only seed 123 correct:",
      ((merged.correct_42 == 0) &
       (merged.correct_123 == 1)).sum())

class_results.to_csv(
    base / "CAFRNet_Seed_Classwise_Diagnostic.csv",
    index=False
)

print("\nSaved class-wise diagnostic.")
```

```python
import os
import json
import shutil
import pandas as pd
from pathlib import Path
from sklearn.model_selection import train_test_split
from torchvision.datasets import OxfordIIITPet
from collections import Counter

ROOT = Path("/content/oxford_pet")
OUT = Path("/content/pet_cafr_data")

# Load the already downloaded official dataset
trainval = OxfordIIITPet(
    root=str(ROOT),
    split="trainval",
    target_types="category",
    download=False
)

test = OxfordIIITPet(
    root=str(ROOT),
    split="test",
    target_types="category",
    download=False
)

classes = trainval.classes
labels = trainval._labels
image_paths = trainval._images

indices = list(range(len(trainval)))

train_idx, val_idx = train_test_split(
    indices,
    test_size=0.20,
    random_state=42,
    stratify=labels
)

# Recreate output structure
if OUT.exists():
    shutil.rmtree(OUT)

for split in ["train", "val", "test"]:
    for class_name in classes:
        (OUT / split / class_name).mkdir(
            parents=True, exist_ok=True
        )

manifest = []

def add_images(dataset, selected_indices, split):
    for idx in selected_indices:
        source = Path(dataset._images[idx])
        label = int(dataset._labels[idx])
        class_name = classes[label]

        destination = OUT / split / class_name / source.name

        # Symlink avoids duplicating image files
        destination.symlink_to(source.resolve())

        manifest.append({
            "split": split,
            "class_index": label,
            "class_name": class_name,
            "source_path": str(source.resolve()),
            "relative_path": str(destination.relative_to(OUT))
        })

add_images(trainval, train_idx, "train")
add_images(trainval, val_idx, "val")
add_images(test, range(len(test)), "test")

manifest_df = pd.DataFrame(manifest)
manifest_df.to_csv(OUT / "split_manifest.csv", index=False)

metadata = {
    "dataset": "Oxford-IIIT Pet",
    "classes": classes,
    "train": len(train_idx),
    "validation": len(val_idx),
    "official_test": len(test),
    "total": len(manifest_df),
    "split_seed": 42,
    "validation_fraction": 0.20,
    "test_source": "Official Oxford-IIIT Pet test split"
}

(OUT / "split_metadata.json").write_text(
    json.dumps(metadata, indent=2)
)

print("===== SPLIT VERIFICATION =====")
for split in ["train", "val", "test"]:
    count = sum(
        1 for p in (OUT / split).rglob("*")
        if p.is_file() or p.is_symlink()
    )
    print(split, count)

print("\nClasses:", len(classes))
print("Manifest rows:", len(manifest_df))
print("Saved:", OUT / "split_manifest.csv")
print("Saved:", OUT / "split_metadata.json")

assert len(train_idx) == 2944
assert len(val_idx) == 736
assert len(test) == 3669
assert manifest_df.groupby("split")["source_path"].nunique().sum() == 7349

print("\nSUCCESS: Split created and verified.")
```

```python
import cv2
import pandas as pd
from pathlib import Path
from PIL import Image
from tqdm.auto import tqdm

ROOT = Path("/content/pet_cafr_data")
OUT = Path("/content/pet_cafr_eval/opencv")

manifest = pd.read_csv(ROOT / "split_manifest.csv")
test_rows = manifest[manifest["split"] == "test"]

created = 0
failed = []

for row in tqdm(test_rows.itertuples(index=False), total=len(test_rows)):
    src = Path(row.source_path)
    dst = OUT / row.class_name / src.name
    dst.parent.mkdir(parents=True, exist_ok=True)

    try:
        image = cv2.imread(str(src))
        if image is None:
            failed.append(str(src))
            continue

        cartoon = cv2.stylization(
            image,
            sigma_s=60,
            sigma_r=0.45
        )

        if not cv2.imwrite(str(dst), cartoon):
            failed.append(str(src))
            continue

        created += 1

    except Exception:
        failed.append(str(src))

print("\n===== OPENCV PET TEST GENERATION =====")
print("Expected:", len(test_rows))
print("Created:", created)
print("Failed:", len(failed))
print("Output:", OUT)

assert created == 3669, "Some images were not generated!"
assert len(failed) == 0, "Check failed image paths!"

print("\nSUCCESS: All 3,669 matched OpenCV test images generated.")
```

```python
import pandas as pd
import numpy as np
from pathlib import Path
from sklearn.metrics import (
    accuracy_score,
    precision_recall_fscore_support,
    classification_report,
    confusion_matrix
)
from scipy.stats import binomtest

RUN = Path("/content/cafr_runs/cafr_v2_oxford_pet_seed42")
CSV = RUN / "pet_test_predictions.csv"

df = pd.read_csv(CSV)

# Expected conditions and matched predictions
assert len(df) == 7338
assert set(df["condition"]) == {"original", "opencv"}

original = df[df.condition == "original"].copy()
opencv = df[df.condition == "opencv"].copy()

# Sort both conditions by relative path to ensure matching
original = original.sort_values("relative_path").reset_index(drop=True)
opencv = opencv.sort_values("relative_path").reset_index(drop=True)

assert original["relative_path"].equals(opencv["relative_path"])
assert original["true_label"].equals(opencv["true_label"])

# 1. Overall metrics
summary = []

for name, part in [("Original", original), ("OpenCV", opencv)]:
    y_true = part["true_label"]
    y_pred = part["predicted_label"]

    precision, recall, f1, _ = precision_recall_fscore_support(
        y_true, y_pred, average="macro", zero_division=0
    )

    summary.append({
        "Condition": name,
        "Images": len(part),
        "Accuracy": accuracy_score(y_true, y_pred),
        "Macro_Precision": precision,
        "Macro_Recall": recall,
        "Macro_F1": f1,
        "Mean_Confidence": part["confidence"].mean()
    })

summary_df = pd.DataFrame(summary)

print("===== OVERALL TEST RESULTS =====")
display(summary_df.round(5))

# 2. Expected Calibration Error (15 bins)
def calculate_ece(part, bins=15):
    confidence = part["confidence"].to_numpy()
    correct = part["correct"].astype(bool).to_numpy()

    edges = np.linspace(0, 1, bins + 1)
    ece = 0.0

    for i in range(bins):
        if i == bins - 1:
            mask = (confidence >= edges[i]) & (confidence <= edges[i+1])
        else:
            mask = (confidence >= edges[i]) & (confidence < edges[i+1])

        if mask.sum():
            ece += mask.mean() * abs(
                correct[mask].mean() - confidence[mask].mean()
            )

    return ece

print("\n===== CALIBRATION =====")
for name, part in [("Original", original), ("OpenCV", opencv)]:
    print(name, "ECE:", round(calculate_ece(part), 6))

# 3. Paired prediction comparison
both_correct = (
    original["correct"].astype(bool) &
    opencv["correct"].astype(bool)
).sum()

original_only = (
    original["correct"].astype(bool) &
    ~opencv["correct"].astype(bool)
).sum()

opencv_only = (
    ~original["correct"].astype(bool) &
    opencv["correct"].astype(bool)
).sum()

both_wrong = (
    ~original["correct"].astype(bool) &
    ~opencv["correct"].astype(bool)
).sum()

print("\n===== PAIRED CORRECTNESS COMPARISON =====")
print("Both correct:", both_correct)
print("Original only correct:", original_only)
print("OpenCV only correct:", opencv_only)
print("Both wrong:", both_wrong)

discordant = original_only + opencv_only

if discordant:
    p_value = binomtest(
        min(original_only, opencv_only),
        discordant,
        p=0.5,
        alternative="two-sided"
    ).pvalue
else:
    p_value = 1.0

print("Exact McNemar-equivalent binomial p-value:", p_value)

# 4. Per-class performance
class_rows = []

for condition, part in [
    ("Original", original),
    ("OpenCV", opencv)
]:
    report = classification_report(
        part["true_label"],
        part["predicted_label"],
        output_dict=True,
        zero_division=0
    )

    for label, metrics in report.items():
        if label not in ["accuracy", "macro avg", "weighted avg"]:
            class_rows.append({
                "Condition": condition,
                "Class": label,
                "Precision": metrics["precision"],
                "Recall": metrics["recall"],
                "F1": metrics["f1-score"],
                "Support": metrics["support"]
            })

class_df = pd.DataFrame(class_rows)

# 5. Save outputs
summary_df.to_csv(RUN / "pet_test_summary.csv", index=False)
class_df.to_csv(RUN / "pet_per_class_metrics.csv", index=False)

original_cm = confusion_matrix(
    original["true_label"],
    original["predicted_label"],
    labels=sorted(original["true_label"].unique())
)

opencv_cm = confusion_matrix(
    opencv["true_label"],
    opencv["predicted_label"],
    labels=sorted(opencv["true_label"].unique())
)

pd.DataFrame(original_cm).to_csv(
    RUN / "pet_original_confusion_matrix.csv", index=False
)
pd.DataFrame(opencv_cm).to_csv(
    RUN / "pet_opencv_confusion_matrix.csv", index=False
)

print("\n===== SAVED ANALYSIS FILES =====")
for filename in [
    "pet_test_summary.csv",
    "pet_per_class_metrics.csv",
    "pet_original_confusion_matrix.csv",
    "pet_opencv_confusion_matrix.csv"
]:
    print(RUN / filename)

print("\nSUCCESS: Oxford-IIIT Pet test analysis completed.")
```

```python
import pandas as pd
import numpy as np
from sklearn.metrics import accuracy_score, f1_score

RUN = "/content/cafr_runs/cafr_v2_oxford_pet_seed42"
df = pd.read_csv(f"{RUN}/pet_test_predictions.csv")

original = (
    df[df.condition == "original"]
    .sort_values("relative_path")
    .reset_index(drop=True)
)
opencv = (
    df[df.condition == "opencv"]
    .sort_values("relative_path")
    .reset_index(drop=True)
)

assert original.relative_path.equals(opencv.relative_path)
assert original.true_label.equals(opencv.true_label)

y = original.true_label.to_numpy()
pred_o = original.predicted_label.to_numpy()
pred_c = opencv.predicted_label.to_numpy()

rng = np.random.default_rng(2026)
N = len(y)
B = 5000

acc_diffs = []
f1_diffs = []

for _ in range(B):
    idx = rng.integers(0, N, size=N)

    acc_o = accuracy_score(y[idx], pred_o[idx])
    acc_c = accuracy_score(y[idx], pred_c[idx])

    f1_o = f1_score(
        y[idx], pred_o[idx], average="macro", zero_division=0
    )
    f1_c = f1_score(
        y[idx], pred_c[idx], average="macro", zero_division=0
    )

    acc_diffs.append(acc_o - acc_c)
    f1_diffs.append(f1_o - f1_c)

def report_ci(values, name):
    low, high = np.percentile(values, [2.5, 97.5])
    print(
        f"{name}: mean={np.mean(values):.5f}, "
        f"95% paired bootstrap CI=[{low:.5f}, {high:.5f}]"
    )

print("===== PAIRED BOOTSTRAP RESULTS =====")
report_ci(acc_diffs, "Accuracy difference (Original - OpenCV)")
report_ci(f1_diffs, "Macro-F1 difference (Original - OpenCV)")

results = pd.DataFrame({
    "accuracy_difference": acc_diffs,
    "macro_f1_difference": f1_diffs
})
results.to_csv(f"{RUN}/pet_paired_bootstrap_5000.csv", index=False)

print("\nSaved:", f"{RUN}/pet_paired_bootstrap_5000.csv")
```

```python
from pathlib import Path
import json
import pandas as pd
import hashlib
import subprocess

REPO = Path("/content/cartoonization-robustness-study")
PKG = REPO / "research/CAFRNet_v2"

# 1. Verify package inventory and hashes
inventory = json.loads(
    (PKG / "package_inventory.json").read_text()
)

errors = []
for item in inventory:
    p = PKG / item["path"]
    if not p.exists():
        errors.append(f"Missing: {item['path']}")
        continue

    digest = hashlib.sha256(p.read_bytes()).hexdigest()
    if digest != item["sha256"]:
        errors.append(f"Hash mismatch: {item['path']}")

# 2. Verify prediction counts
checks = {
    "Imagenette seed42": (
        PKG / "results/imagenette/seed42/test_predictions.csv", 15700
    ),
    "Imagenette seed123": (
        PKG / "results/imagenette/seed123/test_predictions.csv", 15700
    ),
    "Imagenette seed2026": (
        PKG / "results/imagenette/seed2026/test_predictions.csv", 15700
    ),
    "Oxford Pet": (
        PKG / "results/oxford_pet/pet_test_predictions.csv", 7338
    ),
}

print("===== PREDICTION FILE CHECKS =====")
for name, (path, expected) in checks.items():
    if path.exists():
        actual = len(pd.read_csv(path))
        print(name, actual, "/", expected)
        if actual != expected:
            errors.append(f"{name}: expected {expected}, got {actual}")
    else:
        errors.append(f"Missing prediction file: {path}")

# 3. Confirm source snapshot matches active source
for source, snapshot in [
    ("models/cafr_net.py", "source/cafr_net.py"),
    ("training/train_cafr_net.py", "source/train_cafr_net.py"),
    ("evaluation/evaluate_cafr_net.py", "source/evaluate_cafr_net.py"),
]:
    a = (REPO / source).read_bytes()
    b = (PKG / snapshot).read_bytes()
    print("Source match:", source, a == b)
    if a != b:
        errors.append(f"Source snapshot differs: {source}")

# 4. Git status
print("\n===== GIT STATUS =====")
r = subprocess.run(
    ["git", "-C", str(REPO), "status", "--short", "--branch"],
    capture_output=True, text=True
)
print(r.stdout)

print("\n===== AUDIT RESULT =====")
if errors:
    print("AUDIT NEEDS ATTENTION")
    for e in errors:
        print("-", e)
else:
    print("SUCCESS: Package integrity and prediction counts verified.")
```

```python
import subprocess
import getpass
import base64

REPO = "/content/cartoonization-robustness-study"
BRANCH = "research/cafr-net-phase-iv"

def git(args):
    return subprocess.run(
        ["git", "-C", REPO, *args],
        capture_output=True, text=True
    )

# Verify branch
assert git(["branch", "--show-current"]).stdout.strip() == BRANCH

# Verify staged files
staged = git(["diff", "--cached", "--name-only"]).stdout.splitlines()
assert staged, "No staged files found!"

print(f"Files ready to commit: {len(staged)}")

# Commit
commit = git(["commit", "-m", "Add CAFR-Net v2 research package and experimental results"])

if commit.returncode != 0:
    raise RuntimeError(commit.stderr)

print(commit.stdout)

# Credentials
username = input("GitHub username: ").strip()
token = getpass.getpass("GitHub token (hidden): ")

auth = base64.b64encode(
    f"{username}:{token}".encode()
).decode()

try:
    push = subprocess.run(
        [
            "git", "-C", REPO,
            "-c", f"http.extraheader=AUTHORIZATION: basic {auth}",
            "push", "origin", BRANCH
        ],
        capture_output=True,
        text=True
    )

    print(push.stdout)

    if push.returncode != 0:
        print(push.stderr)
        raise RuntimeError("GitHub push failed.")

    print("SUCCESS! Your CAFR-Net v2 research package is on GitHub.")

finally:
    token = None
    auth = None

print(git(["status", "--short", "--branch"]).stdout)
```

```python
cafr_section = """
## 3. Phase IV: CAFR-Net v2 — Cartoon-Aware Feature Refinement Network

CAFR-Net v2 investigates feature refinement for image classification under
cartoonization-induced domain shifts.

### Research objectives

- Improve semantic feature representations across visual domains.
- Investigate the contribution of edge information to cartoonized image recognition.
- Evaluate multi-scale semantic feature fusion.
- Study channel-spatial attention for feature refinement.
- Examine generalization on an external image dataset.

### Architecture

1. **ConvNeXt-Tiny backbone:** extracts hierarchical visual representations.
2. **Multi-scale feature projections:** align features from different backbone stages.
3. **Semantic scale fusion:** combines multi-scale semantic information.
4. **Edge refinement:** incorporates Sobel-based edge information.
5. **Channel-spatial attention:** refines informative channels and spatial regions.
6. **Classification head:** generates final class predictions.

### Experimental evaluation

**Imagenette**

- 10-class image classification benchmark.
- 13,394 verified images.
- Training: 8,048 images.
- Validation: 1,421 images.
- Testing: 3,925 images.
- Three independent seeds: 42, 123 and 2026.
- Evaluation domains: Original, OpenCV, AnimeGANv2 and Stable Diffusion.
- Ablations: without edge refinement, semantic fusion and attention.

**Oxford-IIIT Pet**

- 37-class external dataset.
- Single-seed evaluation using seed 42.
- Original and OpenCV-cartoonized test images.

### Three-seed Imagenette results

| Evaluation domain | Mean accuracy |
|---|---:|
| Original | 99.771% |
| OpenCV | 90.403% |
| AnimeGANv2 | 98.735% |
| Stable Diffusion | 94.726% |

These results describe the evaluated Imagenette experiments. They should not
be interpreted as evidence of universal robustness. OpenCV performance also
shows seed sensitivity.

### Research package

- [CAFR-Net v2 documentation](research/CAFRNet_v2/README.md)
- [Consolidated results](research/CAFRNet_v2/RESULTS_SUMMARY.md)
- [Source code](research/CAFRNet_v2/source/)
- [Experimental results](research/CAFRNet_v2/results/)
- [Dataset manifests](research/CAFRNet_v2/manifests/)

### Limitations

- Oxford-IIIT Pet evaluation uses a single seed.
- Component ablations use single-seed experiments.
- Results on Imagenette and Oxford-IIIT Pet must be reported separately.
- Further independent-dataset experiments are needed before making broader
  generalization claims.

"""

anchor = "## 3. Quick research extension: matched compound-corruption pilot"

if "## 3. Phase IV: CAFR-Net v2" in current_readme:
    print("CAFR-Net v2 section already exists. No insertion needed.")

elif anchor not in current_readme:
    raise ValueError("Insertion heading not found. README was not changed.")

else:
    updated_readme = current_readme.replace(
        anchor,
        cafr_section + "## 4. Quick research extension: matched compound-corruption pilot",
        1
    )

    updated_readme = updated_readme.replace(
        "## 4. Research journal structure",
        "## 5. Research journal structure",
        1
    )
    updated_readme = updated_readme.replace(
        "## 5. Reproducibility and publication checklist",
        "## 6. Reproducibility and publication checklist",
        1
    )
    updated_readme = updated_readme.replace(
        "## 6. Scope and claim discipline",
        "## 7. Scope and claim discipline",
        1
    )

    print("README prepared successfully!")
    print("\nNew section headings:")
    for line in updated_readme.splitlines():
        if line.startswith("## "):
            print(line)
```

```python
import json
import os

folder = "/content/drive/MyDrive/Colab Notebooks"

for filename in os.listdir(folder):
    if filename.endswith(".ipynb"):
        path = os.path.join(folder, filename)

        with open(path, "r", encoding="utf-8") as f:
            nb = json.load(f)

        print("\n" + "=" * 60)
        print("NOTEBOOK:", filename)
        print("Total cells:", len(nb.get("cells", [])))

        output_count = 0

        for i, cell in enumerate(nb.get("cells", [])):
            source = "".join(cell.get("source", []))
            outputs = cell.get("outputs", [])

            output_count += len(outputs)

            keywords = [
                "torch.save", "torch.load", "checkpoint",
                "model.save", "results", "history",
                "CAFR", "train", "evaluate", "drive.mount"
            ]

            if any(word.lower() in source.lower() for word in keywords):
                print(f"\nCell {i}:")
                print(source[:700])

            if outputs:
                print(f"Cell {i} has {len(outputs)} saved output(s)")

        print("\nTotal saved cell outputs:", output_count)
```
